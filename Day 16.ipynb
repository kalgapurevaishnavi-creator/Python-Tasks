```python
#Task1.Create a list original = [1, 2, [3, 4], 5]. Make a shallow copy named shallow and change
#shallow[2][0] = 99. Print both lists and explain the result.

from copy import copy

lists1=[1,2,[3,4],5]
lists2=copy(lists1)

lists2=[2][0]=99

print("oginginal object:", lists1)
print("copy object:", lists2)

```

```python
#2.Do the same as above but use deepcopy instead of shallow copy. Show the difference.

from copy import deepcopy

lists1=[1,2,[3,4],5]
lists2=deepcopy(lists1)

lists2=[2][0]=99

print("original :",lists1)
print("deep:",lists2)
```

```python
#8.Given this data: data = [10, [20, 30], {"name": "Ram"}, 40] Make shallow and deep copies.
#Then change data[1][0] and data[2]["name"]. Print all three versions.

import copy

data = [10, [20, 30], {"name": "Ram"}, 40]

shallow_copy = copy.copy(data)
deep_copy = copy.deepcopy(data)

# Nested list aur dictionary ko change karna
data[1][0] = 100
data[2]["name"] = "Vaishu"

print("Original:", data)
print("Shallow copy:", shallow_copy)
print("Deep copy:", deep_copy)
```

```python
#9(Hard) Create a function that detects whether a copy is shallow or deep by modifying a
#nested element and checking the original.

import copy

def check_copy(copy_type):
    original = [[1, 2, 3], {"name": "ram"}, 40]

    if copy_type == "shallow":
        copied = copy.copy(original)
    else:
        copied = copy.deepcopy(original)

    copied[0][0] = 100

    if original[0][0] == 100:
        print("This is a shallow copy")
    else:
        print("This is a deep copy")


check_copy("shallow")
check_copy("deep")
            
```

```python
#10.create a class Student with attributes name and marks (list). Demonstrate what
#happens when you do shallow copy vs deep copy of a Student object.

import copy
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):
        return f"Name: {self.name}, Marks: {self.marks}"


student1 = Student("Ram", [80, 85, 90])

# Shallow copy
student_shallow = copy.copy(student1)

# Deep copy
student_deep = copy.deepcopy(student1)

# Nested list me changes
student_shallow.marks[0] = 50
student_deep.marks[1] = 40

print("Original student:", student1)
print("Shallow copy:", student_shallow)
print("Deep copy:", student_deep)
```

```python
#13.Mobile
#Create a Mobile class.
#Attributes
#• brand
#• model
#• price
#Create a method that applies a 10% discount and updates price.

class Mobile:
    def __init__(self, brand, model, price):
        self.brand = brand
        self.model = model
        self.price = price

    def apply_discount(self):
        discount = self.price * 10 / 100
        self.price = self.price - discount

    def display(self):
        print("Brand:", self.brand)
        print("Model:", self.model)
        print("Price:", self.price)

mobile1 = Mobile("Samsung", "Galaxy S24", 500)

print("Before discount:")
mobile1.display()

mobile1.apply_discount()

print("\nAfter 10% discount:")
mobile1.display()    
```

```python
#14.Laptop
#Create a Laptop class.
#Attributes
#• brand
#• ram
#• storage
#Create a method to upgrade RAM by 8 GB.

class Laptop:

    def __init__(self, brand, ram, storage):
        self.brand = brand
        self.ram = ram
        self.storage = storage

    def upgrade(self):
        self.ram = self.ram + 8

    def display(self):
        print("Brand:", self.brand)
        print("RAM:", self.ram, "GB")
        print("Storage:", self.storage)


laptop1 = Laptop("Dell", 8, "512GB")

print("Before upgrade:")
laptop1.display()

laptop1.upgrade()

print("\nAfter upgrade:")
laptop1.display()           
```

```python
#15.Bank Account
#Attributes
#• account_holder
#• balance
#Methods
#• deposit()
#• withdraw()
#Withdraw should happen only if balance is sufficient.

class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount
        print("Amount deposited successfully.")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance = self.balance - amount
            print("Amount withdrawn successfully.")
        else:
            print("Insufficient balance.")

    def display(self):
        print("Account Holder:", self.account_holder)
        print("Balance:", self.balance)


account1 = BankAccount("Ram", 1000)

account1.deposit(500)
account1.withdraw(300)
account1.withdraw(1500)

account1.display()
        
```
