```python
#10.Create a class Student with private attribute __marks. Add getter and setter with
#validation (0–100).

class Student:
    def __init__(self):
        self.__marks = 0   # Private attribute

    # Getter
    def get_marks(self):
        return self.__marks

    # Setter with validation
    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks! Enter marks between 0 and 100.")
# Create object
s = Student()

# Set marks
s.set_marks(85)

# Get marks
print("Marks:", s.get_marks())

# Invalid marks
s.set_marks(120)
```

```python
#11.Create class BankAccount with private __balance. Add deposit and withdraw
#methods. Raise custom exception if balance becomes negative.

# Custom Exception
class InsufficientBalanceError(Exception):
    pass

# BankAccount Class
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # Private attribute

    def deposit(self, amount):
        self.__balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if self.__balance - amount < 0:
            raise InsufficientBalanceError("Insufficient Balance!")
        self.__balance -= amount
        print("Withdrawn:", amount)

    def get_balance(self):
        return self.__balance
# Create object
account = BankAccount(5000)

try:
    account.deposit(1000)
    account.withdraw(2000)
    print("Balance:", account.get_balance())

    account.withdraw(5000)   # Raises exception

except InsufficientBalanceError as e:
    print(e)
```

```python
#13.Create class Product with private __price. Add getter and setter. Setter should
#not allow price less than 100.

# Product Class
class Product:
    def __init__(self):
        self.__price = 100   # Private attribute

    # Getter
    def get_price(self):
        return self.__price

    # Setter with validation
    def set_price(self, price):
        if price >= 100:
            self.__price = price
        else:
            print("Price should not be less than 100.")

# Create object
p = Product()

# Set valid price
p.set_price(500)
print("Price:", p.get_price())

# Set invalid price
p.set_price(50)
print("Price:", p.get_price())
```

```python
#14.Create class Rectangle with private length and width. Add method to calculate
#area and perimeter.

# Rectangle Class
class Rectangle:
    def __init__(self, length, width):
        self.__length = length    # Private attribute
        self.__width = width      # Private attribute

    # Method to calculate area
    def area(self):
        return self.__length * self.__width

    # Method to calculate perimeter
    def perimeter(self):
        return 2 * (self.__length + self.__width)

# Create object
r = Rectangle(10, 5)

# Display results
print("Area:", r.area())
print("Perimeter:", r.perimeter())
```

```python
#15.Create single level inheritance: Person → Student. Use super() in constructor.

# Parent Class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Child Class
class Student(Person):
    def __init__(self, name, age, course):
        super().__init__(name, age)   # Call parent constructor
        self.course = course

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Course:", self.course)

# Create object
s = Student("Vaishnavi", 21, "BCA")

# Display details
s.display()
```

```python
#16.Create multi-level inheritance: Person → Employee → Manager. Override a
#method at each level.

# Parent Class
class Person:
    def show(self):
        print("I am a Person")

# Child Class
class Employee(Person):
    def show(self):
        print("I am an Employee")

# Grandchild Class
class Manager(Employee):
    def show(self):
        print("I am a Manager")

# Create object
m = Manager()

# Call overridden method
m.show()
```

```python
#17.Create hierarchical inheritance: Parent Animal → Dog and Cat. Override sound()
#method.

# Parent Class
class Animal:
    def sound(self):
        print("Animal makes a sound")

# Child Class 1
class Dog(Animal):
    def sound(self):
        print("Dog barks....!!")

# Child Class 2
class Cat(Animal):
    def sound(self):
        print("Cat meows.......!!")

# Create objects
dog = Dog()
cat = Cat()

# Call overridden methods
dog.sound()
cat.sound()
```

```python
#18.Create abstract class Shape with abstract method area(). Implement it in Circle
#and Square using super().

from abc import ABC, abstractmethod

# Abstract Class
class Shape(ABC):
    def __init__(self, value):
        self.value = value

    @abstractmethod
    def area(self):
        pass

# Circle Class
class Circle(Shape):
    def __init__(self, radius):
        super().__init__(radius)

    def area(self):
        return 3.14 * self.value * self.value

# Square Class
class Square(Shape):
    def __init__(self, side):
        super().__init__(side)

    def area(self):
        return self.value * self.value

# Create objects
c = Circle(7)
s = Square(5)

print("Circle Area:", c.area())
print("Square Area:", s.area())
```

```python
#19.Create multiple inheritance with two classes Father and Mother. Create child
#class and resolve method conflict.


# Parent Class 1
class Father:
    def show(self):
        print("This is Father's method")

# Parent Class 2
class Mother:
    def show(self):
        print("This is Mother's method")

# Child Class
class Child(Father, Mother):
    def show(self):
        super().show()    # Calls Father's method (MRO)
        print("This is Child's method")

# Create object
c = Child()
c.show()
```

```python
#20.Create a class Calculator that can add 2 or 3 numbers (method overloading
#using *args).

# Calculator Class
class Calculator:

    def add(self, *args):
        if len(args) == 2:
            print("Sum of 2 numbers:", args[0] + args[1])
        elif len(args) == 3:
            print("Sum of 3 numbers:", args[0] + args[1] + args[2])
        else:
            print("Please enter only 2 or 3 numbers.")

# Create object
c = Calculator()

# Add 2 numbers
c.add(10, 20)

# Add 3 numbers
c.add(10, 20, 30)

# Invalid input
c.add(10)
```

```python
#21.Create class Point and overload + and > operators.

# Point Class
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overload + operator
    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    # Overload > operator
    def __gt__(self, other):
        return (self.x + self.y) > (other.x + other.y)

    def display(self):
        print("Point(", self.x, ",", self.y, ")")

# Create objects
p1 = Point(2, 3)
p2 = Point(4, 5)

# + operator
p3 = p1 + p2
print("After Addition:")
p3.display()

# > operator
if p1 > p2:
    print("p1 is greater than p2")
else:
    print("p2 is greater than or equal to p1")
```
