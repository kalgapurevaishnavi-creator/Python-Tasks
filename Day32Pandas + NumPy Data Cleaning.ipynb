```python
#12.Find all unique values in the gender column and convert them into consistent
#categories such as Male and Female.

df['gender'] = (
    df['gender']
    .str.strip()
    .str.lower()
    .replace({
        'm': 'Male',
        'male': 'Male',
        'f': 'Female',
        'female': 'Female'
    })
)

print(df['gender'].unique())                    #check the male female or Nan
df
```

```python
#13.Clean the is_active column so values such as Yes, yes, Y, 1 become 1, and
#No, no, N, 0 become 0.

df['is_active'] = (
    df['is_active']
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({
        'yes': 1,
        'y': 1,
        '1': 1,
        'no': 0,
        'n': 0,
        '0': 0
    })
)
df

df['is_active'].unique()
df
```

```python
#14.Convert the churn_flag column from categorical values (Yes/No) into a binary
#numerical column (1/0).


# Convert churn_flag Yes/No to 1/0
df["churn_flag"] = df["churn_flag"].map({
    "Yes": 1,
    "No": 0
})

print(df["churn_flag"].head())
```

```python
#15.Identify all values in income_usd that cannot be directly interpreted as
#numbers.

# Convert income_usd to numeric
# Invalid values become NaN
income_numeric = pd.to_numeric(df["income_usd"], errors="coerce")

# Display values that cannot be interpreted as numbers
invalid_income = df.loc[
    income_numeric.isna() & df["income_usd"].notna(),
    "income_usd"
]

print(invalid_income)
```

```python
#16.Clean income_usd by removing currency symbols, commas, and invalid text,
#then convert it to a numeric column using pd.to_numeric().

# Remove currency symbols and commas
df["income_usd"] = (
    df["income_usd"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

# Convert invalid text to NaN and make the column numeric
df["income_usd"] = pd.to_numeric(df["income_usd"], errors="coerce")

print(df["income_usd"].head())
print(df["income_usd"].dtype)    
```

```python
#17.Clean the Age column so values such as "35 years" become numeric values.

# Remove "years" and convert Age to numeric
df["age"] = (
    df["age"]
    .astype(str)
    .str.replace("years", "", regex=False)
    .str.strip()
)

df["age"] = pd.to_numeric(df["age"], errors="coerce")

print(df["age"].head())
print(df["age"].dtype)
```

```python
#18.Identify unrealistic age values using NumPy/Pandas and replace them with
#NaN.


import numpy as np

# Identify unrealistic age values
df.loc[(df["age"] < 18) | (df["age"] > 100), "age"] = np.nan

print(df["age"].head())
```

```python
print("Unrealistic age values:",
      df[(df["age"] < 18) | (df["age"] > 100)]["age"].tolist())
```

```python
#19.Clean the credit_score column by removing text such as "pts" and converting
#the column to numeric.

# Remove "pts" and convert credit_score to numeric
df["credit_score"] = (
    df["credit_score"]
    .astype(str)
    .str.replace("pts", "", regex=False)
    .str.strip()
)

df["credit_score"] = pd.to_numeric(
    df["credit_score"],
    errors="coerce"
)
df["credit_score"].head()
df["credit_score"].dtype
```

```python
#20.Detect outliers in income_usd using the IQR method with NumPy/Pandas.

import numpy as np

# Calculate Q1 and Q3
Q1 = df["income_usd"].quantile(0.25)
Q3 = df["income_usd"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate lower and upper limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Find outliers
outliers = df[
    (df["income_usd"] < lower_limit) |
    (df["income_usd"] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

print("\nOutliers:")
print(outliers["income_usd"])
```

```python
#21.Detect outliers in credit_score using the IQR method.

# Calculate Q1 and Q3
Q1 = df["credit_score"].quantile(0.25)
Q3 = df["credit_score"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Find outliers
outliers = df[
    (df["credit_score"] < lower_limit) |
    (df["credit_score"] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

print("\nOutliers:")
print(outliers["credit_score"])
```

```python
#22.Detect suspicious values such as -999 and 99999 in tenure_years and
#satisfaction.

suspicious = df[
    df["tenure_years"].isin([-999, 99999]) |
    df["satisfaction"].isin([-999, 99999])
]

print(suspicious[["tenure_years", "satisfaction"]])
```

```python
#23.Replace invalid/suspicious numerical values with NaN.

import numpy as np

# Replace suspicious values with NaN
df["tenure_years"] = df["tenure_years"].replace([-999, 99999], np.nan)

df["satisfaction"] = df["satisfaction"].replace([-999, 99999], np.nan)

# Check again
print(df[["tenure_years", "satisfaction"]].head())
```

```python
df["tenure_years"].isin([-999, 99999]).sum()
df["satisfaction"].isin([-999, 99999]).sum()
```

```python
#24.Fill missing numerical values using the median of each numerical column.

# Select numerical columns
numeric_cols = df.select_dtypes(include=np.number).columns

# Fill missing values with column-wise median
df[numeric_cols] = df[numeric_cols].fillna(
    df[numeric_cols].median()
)

# Check missing values
df[numeric_cols].isnull().sum()
```

```python
#25.Fill missing categorical values using the mode of each categorical column.
# Select categorical columns
categorical_cols = df.select_dtypes(include="object").columns

# Fill missing categorical values with the mode of each column
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

# Check missing values
df[categorical_cols].isnull().sum()
```

```python
#26.Clean the signup_date column, which contains multiple date formats and
#invalid dates, and convert it to Pandas datetime.

# Convert signup_date to datetime
df["signup_date"] = pd.to_datetime(
    df["signup_date"],
    errors="coerce"
)
df
# Check the result
df["signup_date"].head()
df["signup_date"].dtype
df
```

```python
#27.Find rows where signup_date could not be converted into a valid date.

# Find rows where signup_date is invalid
invalid_dates = df[df["signup_date"].isna()]

print(invalid_dates)
```

```python
#28.Clean the lead_source column by standardizing values such as Website,
#website, WEB, Referral, referral , Google Ads, and google_ads.


#Clean and standardize lead_source
df["lead_source"] = (
    df["lead_source"]
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({
        "web": "website",
        "website": "website",
        "referral": "referral",
        "google_ads": "google ads",
        "google ads": "google ads"
    })
)

# Check unique values
df["lead_source"].unique()
```

```python
#29.Create a final cleaned DataFrame containing:
#• cleaned column names
#• no duplicate IDs
#• standardized categorical values
#• numeric columns converted to numeric types
#• invalid values converted to NaN
#• missing values handled
#• detected outliers handled


# Make a copy of the cleaned dataset
final_df = df.copy()

# 1. Clean column names
final_df.columns = (
    final_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# 2. Remove duplicate Customer IDs
final_df = final_df.drop_duplicates(subset="customer_id")

# 3. Convert numeric columns to numeric
numeric_cols = [
    "age",
    "income_usd",
    "credit_score",
    "tenure_years",
    "monthly_visits",
    "satisfaction",
    "churn_flag"
]
for col in numeric_cols:
    final_df[col] = pd.to_numeric(final_df[col], errors="coerce")

# 4. Replace invalid/suspicious values with NaN
final_df["tenure_years"] = final_df["tenure_years"].replace([-999, 99999], np.nan)
final_df["satisfaction"] = final_df["satisfaction"].replace([-999, 99999], np.nan)

# 5. Fill numerical missing values with median
final_df[numeric_cols] = final_df[numeric_cols].fillna(
    final_df[numeric_cols].median()
)

# 6. Fill categorical missing values with mode
categorical_cols = final_df.select_dtypes(include="object").columns

for col in categorical_cols:
    final_df[col] = final_df[col].fillna(final_df[col].mode()[0])

# 7. Convert signup_date to datetime
final_df["signup_date"] = pd.to_datetime(
    final_df["signup_date"],
    errors="coerce"
)

# 8. Check final DataFrame
print(final_df.head())
print("\nShape:", final_df.shape)
print("\nData Types:")
print(final_df.dtypes)
print("\nMissing Values:")
print(final_df.isnull().sum())
```
