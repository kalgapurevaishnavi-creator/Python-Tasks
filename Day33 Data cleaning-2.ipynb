```python
#Data Type & Invalid Values
```

```python
#11.The Age column contains numbers as well as values like "unknown" and
#"twenty five". Clean this column and convert it into a proper numeric column.

df['age'] = df['age'].replace({
    'unknown': pd.NA,
    'twenty five': 25
})

df['age'] = pd.to_numeric(df['age'], errors='coerce')

print(df['age'])
```

```python
#12.The Annual Income column contains values such as 650000 INR, ₹720000,
#1.2L, and N/A. Clean the column and convert the valid values into numeric
#format.

import numpy as np
import pandas as pd

def clean_income(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip().upper()

    # N/A or unknown values
    if value in ['N/A', 'NA', 'UNKNOWN', 'NONE', '']:
        return np.nan

    # Remove currency symbols and INR
    value = value.replace('₹', '').replace('INR', '').strip()

    # Values like 1.2L = 120000
    if value.endswith('L'):
        return float(value[:-1]) * 100000

    # Normal numeric values
    try:
        return float(value)
    except:
        return np.nan

df['annual_income'] = df['annual_income'].apply(clean_income)

df['annual_income']
```

```python
#13.Clean the Credit Score column and identify values that cannot be converted
#into numbers.

#Clean Credit Score column
df['credit_score'] = pd.to_numeric(df['credit_score'], errors='coerce')

# Identify values that cannot be converted
invalid_values = df[df['credit_score'].isna()]

print(invalid_values)
```

```python
#14.The Loan Amount column contains currency symbols, text, and numeric
#values. Remove the unwanted characters and convert it into numeric format.


# Remove unwanted characters and convert Loan Amount to numeric
df['loan_amount_(inr)'] = (
    df['loan_amount_(inr)']
    .astype(str)
    .str.replace(r'[₹$,]', '', regex=True)
    .str.strip()
)

df['loan_amount_(inr)'] = pd.to_numeric(df['loan_amount_(inr)'], errors='coerce')

print(df['loan_amount_(inr)'])
```

```python
#15.The Monthly Visits column contains values such as "five", "N/A", and "-".
#Convert valid values into numeric values and handle invalid values.



# Convert Monthly Visits into numeric values
df['monthly_visits'] = pd.to_numeric(
    df['monthly_visits'].replace({
        'five': 5,
        'N/A': np.nan,
        '-': np.nan
    }),
    errors='coerce'
)

df['monthly_visits']
```

```python
#16.Clean the Debt to Income column where some values are written like 35%
#and 0.4x. Convert the valid values into a consistent numeric format.

def clean_dti(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    if value in ['unknown', 'n/a', 'na', '-']:
        return np.nan

    if value.endswith('%'):
        return float(value[:-1]) / 100

    if value.endswith('x'):
        return float(value[:-1])

    try:
        return float(value)
    except:
        return np.nan

df['debt_to_income'] = df['debt_to_income'].apply(clean_dti)

df['debt_to_income']
```

```python
#17.After cleaning the numerical columns, check which columns are still stored as
#object data type.

#Check columns that are still object datatype
df.select_dtypes(include='object').columns
```

```python
#Categorical Cleaning
```

```python
#18.Find all unique values in the Employment-Type column.

df['employment-type'].unique()
```

```python
#19.Standardize the Employment-Type column so that values such as salaried,
#SALARIED, and Salaried become one consistent value.

df['employment-type'] = (
    df['employment-type']
    .str.strip()
    .str.lower()
    .replace({
        'self employed': 'self-employed',
        'unknown': np.nan
    })
)

df['employment-type'].unique()
```

```python
#20.Clean the Education Level column and make values such as graduate,
#GRADUATE, and Graduate consistent.


df['education__level'] = (
    df['education__level']
    .str.strip()
    .str.lower()
)

df['education__level'].unique()
```

```python
#21.Clean the City column. Identify incorrect or inconsistent values such as
#hyderbad, Hyd, Bangalore, and Bengaluru.

#Clean City column

df['city'] = (
    df['city']
    .str.strip()
    .str.lower()
    .replace({
        'hyderbad': 'hyderabad',
        'hyd': 'hyderabad',
        'bangalore': 'bengaluru'
    })
)

df['city'].unique() 
```

```python
#22.Find all unique values in the Customer Segment column and identify invalid
#categories.

#Find unique values in Customer Segment
df['customer_segment'].unique()


# Identify invalid categories
valid_segments = ['retail', 'sme', 'corporate', 'premium']

invalid_segments = df[
    ~df['customer_segment'].str.strip().str.lower().isin(valid_segments)
    & df['customer_segment'].notna()
]

invalid_segments['customer_segment'].unique()
```

```python
#23.Clean the Loan Type column so that values such as personal, PERSONAL,
#and personal loan are standardized.

#Clean Loan Type column
df['loan_type'] = (
    df['loan_type']
    .str.strip()
    .str.lower()
    .replace({
        'personal loan': 'personal'
    })
)

df['loan_type'].unique()
```

```python
#24.Clean the Loan Approved column where values include Yes, yes, Y, 1, No,
#NO, N, and 0. Convert them into only Yes/No.

#Clean Loan Approved column
df['loan_approved?'] = (
    df['loan_approved?']
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({
        'yes': 'Yes',
        'y': 'Yes',
        '1': 'Yes',
        'no': 'No',
        'n': 'No',
        '0': 'No',
        'nan': np.nan
    })
)

df['loan_approved?'].unique()
```

```python
#27.Detect outliers in Annual Income using the IQR method.

#Detect outliers in Annual Income using IQR

Q1 = df['annual_income'].quantile(0.25)
Q3 = df['annual_income'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df['annual_income'] < lower_limit) |
    (df['annual_income'] > upper_limit)
]

print(outliers[['annual_income']])
print("Number of outliers:", len(outliers))
```

```python
#28.Detect outliers in Loan Amount using the IQR method.

#Detect outliers in Loan Amount using IQR
Q1 = df['loan_amount_(inr)'].quantile(0.25)
Q3 = df['loan_amount_(inr)'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df['loan_amount_(inr)'] < lower_limit) |
    (df['loan_amount_(inr)'] > upper_limit)
]

outliers[['loan_amount_(inr)']]
len(outliers)
```

```python
#29.Check the Debt to Income column and identify impossible values such as
#negative values or values greater than 1.


#Identify impossible Debt to Income values
invalid_dti = df[
    (df['debt_to_income'] < 0) |
    (df['debt_to_income'] > 1)
]

invalid_dti[['debt_to_income']]
```

```python
#30.Final cleaning task: Create a completely cleaned version of the dataset by:
#• Cleaning column names
#• Removing duplicates
#• Handling missing values
#• Fixing data types
#• Cleaning numerical columns
#• Standardizing categorical values
#• Handling invalid values
#• Detecting and treating outliers
#• Checking the final dataset for remaining errors


#FINAL DATA CLEANING

#Clean column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
)

# Remove completely duplicate rows
df = df.drop_duplicates()

# Remove duplicate Customer IDs
df = df.drop_duplicates(subset='customer_id')

# Clean numerical columns
df['age'] = pd.to_numeric(df['age'], errors='coerce')
df['annual_income'] = pd.to_numeric(df['annual_income'], errors='coerce')
df['credit_score'] = pd.to_numeric(df['credit_score'], errors='coerce')
df['loan_amount_(inr)'] = pd.to_numeric(df['loan_amount_(inr)'], errors='coerce')
df['tenure_months'] = pd.to_numeric(df['tenure_months'], errors='coerce')
df['monthly_visits'] = pd.to_numeric(df['monthly_visits'], errors='coerce')
df['debt_to_income'] = pd.to_numeric(df['debt_to_income'], errors='coerce')

# Handle invalid Age
df.loc[(df['age'] < 0) | (df['age'] > 100), 'age'] = np.nan

# Handle invalid Credit Score
df.loc[
    (df['credit_score'] < 300) | (df['credit_score'] > 850),
    'credit_score'
] = np.nan

# Handle invalid Debt to Income
df.loc[
    (df['debt_to_income'] < 0) | (df['debt_to_income'] > 1),
    'debt_to_income'
] = np.nan

# Standardize Employment Type
df['employment-type'] = (
    df['employment-type']
    .str.strip()
    .str.lower()
    .replace({
        'self employed': 'self-employed',
        'unknown': np.nan
    })
)

#Standardize Education Level
df['education__level'] = (
    df['education__level']
    .str.strip()
    .str.lower()
    .replace({
        'post graduate': 'postgraduate'
    })
)

#Standardize City
df['city'] = (
    df['city']
    .str.strip()
    .str.lower()
    .replace({
        'hyderbad': 'hyderabad',
        'hyd': 'hyderabad',
        'bangalore': 'bengaluru'
    })
)

# 11. Standardize Loan Type
df['loan_type'] = (
    df['loan_type']
    .str.strip()
    .str.lower()
    .replace({
        'personal loan': 'personal'
    })
)

# 12. Standardize Loan Approved
df['loan_approved?'] = (
    df['loan_approved?']
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({
        'yes': 'Yes',
        'y': 'Yes',
        '1': 'Yes',
        'no': 'No',
        'n': 'No',
        '0': 'No',
        'nan': np.nan
    })
)

# Treat Annual Income outliers using IQR
Q1 = df['annual_income'].quantile(0.25)
Q3 = df['annual_income'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df.loc[
    (df['annual_income'] < lower) |
    (df['annual_income'] > upper),
    'annual_income'
] = np.nan

#Treat Loan Amount outliers using IQR
Q1 = df['loan_amount_(inr)'].quantile(0.25)
Q3 = df['loan_amount_(inr)'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df.loc[
    (df['loan_amount_(inr)'] < lower) |
    (df['loan_amount_(inr)'] > upper),
    'loan_amount_(inr)'
] = np.nan

#Fill missing numerical values with median
numeric_columns = [
    'age',
    'annual_income',
    'credit_score',
    'loan_amount_(inr)',
    'tenure_months',
    'monthly_visits',
    'debt_to_income'
]

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

#Fill missing categorical values with mode
categorical_columns = [
    'employment-type',
    'education__level',
    'marital_status',
    'city',
    'customer_segment',
    'loan_type',
    'loan_approved?'
]

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])

#Final data-quality checks
df.isna().sum()
df.duplicated().sum()
df['customer_id'].duplicated().sum()
df.dtypes
df.shape
df.head()
```
