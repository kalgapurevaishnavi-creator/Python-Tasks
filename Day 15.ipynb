# lambda funcation in python

```python
#1.Write a lambda function to return the square of a number.

squared = lambda x : x*x
squared(5)
```

```python
#2.Write a lambda function to return the cube of a number.

cube=lambda x:pow(x,3)
cube(4)
```

```python
#3.Write a lambda function to add two numbers.

add=lambda x,y: x+y
add(10,20)
```

```python
#4.Write a lambda function to find the maximum of two numbers.

max=lambda x,y:x if x>y else y
print(max(15,8))
```

```python
#5.Write a lambda function that returns "Even" if a number is even; otherwise "Odd".



result=(lambda x:"even" if x % 2==0 else "odd")
result(9)
```

```python
#6.Use map() with lambda to square every element of a list.
numbers=[1,2,3,4,5]

square=list(map(lambda x: x**2,numbers ))
print(square)
```

```python
#7.Use filter() with lambda to keep only even numbers.

numbers=[1,2,3,4,5,6,7]

even=list(filter(lambda x:x % 2 == 0,numbers))
print(even)
```

```python
#8.Use filter() with lambda to keep numbers greater than 50.

lists=[20,45,55,75]
num=(filter(lambda x:x>50,lists))
print(55,70)
```

```python
#9.9.Sort a list of numbers in ascending order using lambda.

lists=[2,3,4,5,6,8,10,11,12,13]
result=sorted(lists,key=lambda x:x)
print(result)
```

```python
#10.Sort a list of numbers in descending order using lambda.

lists=[2,3,4,5,6,7,8,9]
result=sorted(lists,key=lambda x:x,reverse=True)
print(result)
```

```python
#11.Use reduce() with lambda to find the sum of all numbers.

from functools import reduce
num=[10,20,30,40]

result =reduce(lambda x,y:x + y,num)
print(result)
```

```python
#12.Use reduce() with lambda to find the product of all numbers.

from functools import reduce
num=[2,3,4,5]
result=reduce(lambda x,y:x*y,num)
print(result)
```
