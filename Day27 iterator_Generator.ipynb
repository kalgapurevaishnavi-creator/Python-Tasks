```python
#6.Problem 6
#Create a custom iterator that prints numbers from 1 to 10.


class NumberIterator:
    def __init__(self):   
        self.num = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.num <= 10:
            value = self.num
            self.num += 1
            return value
        else:
            raise StopIteration

# Create iterator object
numbers = NumberIterator()

# Print numbers from 1 to 10
for i in numbers:
    print(i)   
```

```python
#Problem 8
#Create a custom iterator that prints numbers in reverse order from
#10 to 1.

class ReverseIterator:
    def __init__(self):
        self.num = 10

    def __iter__(self):
        return self

    def __next__(self):
        if self.num >= 1:
            value = self.num
            self.num -= 1
            return value
        else:
            raise StopIteration

# Create iterator object
rev = ReverseIterator()

# Print numbers from 10 to 1
for i in rev:
    print(i)
```

```python
#9.Problem 9
#Create a custom iterator that prints the characters of the word
#"GENERATOR" one by one.

class WordIterator:
    def __init__(self, word):
        self.word = word
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.word):
            ch = self.word[self.index]
            self.index += 1
            return ch
        else:
            raise StopIteration

# Create iterator
obj = WordIterator("GENERATOR")

# Print characters one by one
for i in obj:
    print(i)
```

```python
#10.Problem 10
#Create a custom iterator that prints the multiplication table of 5.
#Expected Output:
#5
#10
#15
#20
#25
#30
#35
#40
#45
#50

class TableIterator:
    def __init__(self, number):
        self.number = number
        self.i = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.i <= 10:
            result = self.number * self.i
            self.i += 1
            return result
        else:
            raise StopIteration

# Create iterator for table of 5
table = TableIterator(5)

# Print the table
for value in table:
    print(value)
```

```python
#16.Problem 16
#Given the following list:
#numbers = [10, 15, 20, 25, 30, 35]
#Create a generator that yields only the even numbers.

# Given list
numbers = [10, 15, 20, 25, 30, 35]

# Generator function
def even_numbers(nums):
    for i in nums:
        if i % 2 == 0:
            yield i

# Create generator object
gen = even_numbers(numbers)

# Print even numbers
for num in gen:
    print(num)
```

```python
#23.Problem 23
#Create a custom iterator that prints all alphabet letters from A to Z.

class AlphabetIterator:
    def __init__(self):
        self.letter = 'A'

    def __iter__(self):
        return self

    def __next__(self):
        if self.letter <= 'Z':
            current = self.letter
            self.letter = chr(ord(self.letter) + 1)
            return current
        else:
            raise StopIteration

# Create iterator object
obj = AlphabetIterator()

# Print A to Z
for i in obj:
    print(i)
```

```python
#24.Problem 24
#Create a custom iterator that prints all vowels from the string below.
#text = "Artificial Intelligence"

class VowelIterator:
    def __init__(self, text):
        self.text = text
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        while self.index < len(self.text):
            ch = self.text[self.index]
            self.index += 1

            if ch.lower() in "aeiou":
                return ch

        raise StopIteration

# Given string
text = "Artificial Intelligence"

# Create iterator object
obj = VowelIterator(text)

# Print vowels
for vowel in obj:
    print(vowel)
```

```python
#25.Problem 25
#Create a custom iterator that prints the Fibonacci sequence up to 10
#terms.

class Fibonacci:
    def __init__(self):
        self.a = 0
        self.b = 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < 10:
            value = self.a
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return value
        else:
            raise StopIteration
# Create iterator object
obj = Fibonacci()

# Print first 10 Fibonacci terms
for i in obj:
    print(i)
```

```python
#26.Problem 26
#Create a generator that yields the Fibonacci sequence up to 15
#terms.

# Generator function
def fibonacci():
    a = 0
    b = 1

    for i in range(15):
        yield a
        a, b = b, a + b

# Create generator object
gen = fibonacci()

# Print Fibonacci sequence
for num in gen:
    print(num)
```

```python
#27.Problem 27
#Create a generator that yields all prime numbers between 1 and 100.

# Generator function
def prime_numbers():
    for num in range(2, 101):
        is_prime = True

        for i in range(2, num):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            yield num

# Create generator object
gen = prime_numbers()

# Print prime numbers
for prime in gen:
    print(prime)
```

```python
#28.Problem 28
#Create a generator that yields all factors of a given number.
#Example Input:
#num = 24

# Generator function
def factors(num):
    for i in range(1, num + 1):
        if num % i == 0:
            yield i

# Example input
num = 24

# Create generator object
gen = factors(num)

# Print factors
for factor in gen:
    print(factor)
```

```python
#29.Problem 29
#Create a generator that yields the countdown from 20 to 1.

# Generator function
def countdown():
    for i in range(20, 0, -1):
        yield i

# Create generator object
gen = countdown()

# Print countdown
for num in gen:
    print(num)
```

```python
#30.Problem 30
#Create an infinite generator that yields multiples of 5.
#Print only the first 15 values using next().

# Infinite generator
def multiples_of_five():
    num = 5
    while True:
        yield num
        num += 5

# Create generator object
gen = multiples_of_five()

# Print first 15 multiples using next()
for i in range(15):
    print(next(gen))
```
