```python
#10.Convert Date_of_Journey into a datetime format.

df["Date_of_Journey"] = pd.to_datetime(
    df["Date_of_Journey"],
    dayfirst=True
)

df["Date_of_Journey"].head()
```

```python
#15.Convert Dep_Time into datetime format and extract the departure hour into
#Dep_Hour.

df["Dep_Time"] = pd.to_datetime(df["Dep_Time"], format="%H:%M")

df["Dep_Hour"] = df["Dep_Time"].dt.hour

df[["Dep_Time", "Dep_Hour"]].head()
```

```python
#16.Extract the departure minute from Dep_Time and create Dep_Minute.

df["Dep_Minute"] = df["Dep_Time"].dt.minute

df[["Dep_Time", "Dep_Hour", "Dep_Minute"]].head()
```

```python
#17.Drop the original Dep_Time column after extracting the required features.

##df = df.drop(columns=["Dep_Time"])       #drop the Dep_columns
##df.head

df.columns
```

```python
#18.Extract only the time portion from Arrival_Time.

df["Arrival_Time"] = df["Arrival_Time"].astype(str).str[:5]

df[["Arrival_Time"]].head()
```

```python
#19.Extract the arrival hour and minute into Arrival_Hour and Arrival_Minute.

df["Arrival_Hour"] = pd.to_datetime(
    df["Arrival_Time"], format="%H:%M"
).dt.hour

df["Arrival_Minute"] = pd.to_datetime(
    df["Arrival_Time"], format="%H:%M"
).dt.minute

df[["Arrival_Time", "Arrival_Hour", "Arrival_Minute"]].head()
```

```python
#20.Drop the original Arrival_Time column after extracting the required features.

df = df.drop(columns=["Arrival_Time"])

df.head()
```

```python
#21.Extract the hour component from Duration and create Duration_Hour. Handle
#invalid or missing extracted values appropriately.

df["Duration_Hour"] = (
    df["Duration"]
    .astype(str)
    .str.extract(r"(\d+)\s*h", expand=False)
)

df["Duration_Hour"] = pd.to_numeric(
    df["Duration_Hour"], errors="coerce"
)

df["Duration_Hour"] = df["Duration_Hour"].fillna(0).astype(int)

df[["Duration", "Duration_Hour"]].head()    
```

```python
#22.Extract the minute component from Duration and create Duration_Minute.
#Handle invalid or missing extracted values appropriately.

df["Duration_Minute"] = (
    df["Duration"]
    .astype(str)
    .str.extract(r"(\d+)\s*h", expand=False)
)

df["Duration_Minute"] = pd.to_numeric(
    df["Duration_Minute"], errors="coerce"
)

df["Duration_Minute"] = df["Duration_Minute"].fillna(0).astype(int)

df[["Duration", "Duration_Minute"]].head()  
```

```python
#23.Drop the original Duration column after creating the numerical duration
#features.

df = df.drop(columns=["Duration"])
df.head()    
```

```python
#24.Replace non-stop in Total_Stops with 0 stops.

df["Total_Stops"] = df["Total_Stops"].replace("non-stop", 0)

df["Total_Stops"].value_counts(dropna=False)
```

```python
#25.Extract only the numerical value from Total_Stops and convert the column into
#an integer data type. Handle missing or invalid values appropriately.

df["Total_Stops"] = (
    df["Total_Stops"]
    .astype(str)
    .str.extract(r"(\d+)", expand=False)
)

df["Total_Stops"] = pd.to_numeric(
    df["Total_Stops"], errors="coerce"
)

df["Total_Stops"] = df["Total_Stops"].fillna(0).astype(int)
df["Total_Stops"].head()
```

```python
#26.Split the Route column using → as the separator and create Route_1,
#Route_2, Route_3, Route_4, and Route_5.

route_parts = df["Route"].str.split("→", expand=True)

df["Route_1"] = route_parts[0]
df["Route_2"] = route_parts[1]
df["Route_3"] = route_parts[2]
df["Route_4"] = route_parts[3]
df["Route_5"] = route_parts[4]

df[["Route_1", "Route_2", "Route_3", "Route_4", "Route_5"]].head()
```

```python
#27.Identify the missing values created in Route_3, Route_4, and Route_5 after
#splitting the route.

df[["Route_3", "Route_4", "Route_5"]].isnull().sum()  
```

```python
#28.Replace the missing values in the route columns with an appropriate value
#such as None.

route_columns = ["Route_1", "Route_2", "Route_3", "Route_4", "Route_5"]

df[route_columns] = df[route_columns].fillna("None")

df[route_columns].head()
```

```python
#29.Drop the original Route column after creating and cleaning the route features.
df = df.drop(columns=["Route"])
df.head()
```
