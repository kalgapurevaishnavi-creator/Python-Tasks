# Multiple inheritance(static method & abstract method


```python
#Task1.Create two parent classes Father and Mother. Create a child class Child that inherits from both.
###############(basic multiple inheritance)#######################

#parent1
class Father:
    def __init__(self,name,**kwargs):
        super().__init__(**kwargs)

        self.name=name
#parent2
class Mother:
    def __init__(self,value,**kwargs):
        super().__init__(**kwargs)

        self.value=value
#child class
class Child(Father,Mother):
    def __init__(self,name,value,classes):
        super().__init__(name=name,
                         value=value)
        self.classes=classes
        
    def display_all_info(self):
        print(f"name :{self.name}")
        print(f"value :{self.value}")
        print(f"classes :{self.classes}")
        
#create a object
c1=Child("Vaishu",22,"DS")
#display all information
c1.display_all_info()
    
```

```python
#Task2.Define class A with method show() and class B with method display(). Create class C that
#inherits from both and calls both methods.


class A:
    def show(self):
        print("show A")
        
class B:
    def display(self):
        print("Display B")
        
class C(A,B):
     def display_all(self):
        self.show()
        self.display()
        
#create a object
c1=C()
#display all information
c1.display_all()
```

```python
#3.Create classes Teacher and Author. Create class Person that inherits from both and has
#attributes from both.

class Teacher:
    def __init__(self,school):
        self.school=school
        
    def display_teacher(self):
        print(f"school :{self.school}")

class Author:
    def __init__(self,book):
        self.book=book
        
    def display_Author(self):
        print(f"book :{self.book}")
        
class Person(Teacher,Author):
    def __init__(self,school,book,name):
        Teacher.__init__(self,school)
        Author.__init__(self,book)
        self.name=name
        
    def display_Person_info(self):
        print(f"name:{self.name}")
        self.display_teacher()
        self.display_Author()
#create a instance of class with attribute
p1=Person(
    school="zpp",
    book="story",
    name="Shruti")
#display all information
p1.display_Person_info()
        
```

```python
#4.Define class Sports and class Academic. Create class Student that inherits from both.

class Sports:
    def Sport_info(self):
        print("play sports")
class Academic:
    def Academic_info(self):
        print("Grade:A")
class Student(Sports,Academic):
    pass

#demonstration
s = Student()
s.Sport_info()
s.Academic_info()
```

```python
#5.Create class Camera and class MusicPlayer. Create class Smartphone that inherits from both.


class Camera:
    def take_photo(self):
        print("open Camera")

class Musicplayer:
    def play_music(self):
        print("play MusicPlayer")
class Smartphone(Camera,Musicplayer):
    pass

#demonstration
phone = Smartphone()
phone.take_photo()
phone.play_music()
```

```python
#6.Define two classes with same method name. Create a child class that inherits from both
#and resolve the method conflict.

#parent1
class A:
    def show(self):
        print("method from class A")
#parent2
class B:
    def show(self):
        print("method from class B")
        
#child class inherits frpm both
class C(A,B):
    def show(self):
        print("method from child class C")
        
#object creation
obj = C()

#child class method is called
obj.show()
```

```python
#7.Create class Engine and class Wheels. Create class Car that inherits from both.

#parent1
class Engine:
    def play(self):
        print("The play Engine")
#parent2
class Wheels:
    def play(self):
        print("The play Wheels")       
#child class inherits from both
class Car(Engine,Wheels):
    def play(self):
        print("The play Car")         
#object creation
c=Car()
#child class method call
c.play()
        
```

```python
#8.Define class Writer and class Singer. Create class Artist that inherits from both.

#parent1
class Writer:
    def show(self):
        print("you are writting")

#parent2
class Singer:
    def show(self):
        print("you are singer")
#child class
class Artist(Writer,Singer):
    def show(self):
        print("you are artist")
        
#create a object
a1=Artist()
#child class method call
a1.show()
```

```python
#9.Create class Bank and class Insurance. Create class Customer that inherits from both.

#parent1
class Bank:
    def pay(self):
        print("pay your money")

#parent2
class Insurance:
    def pay(self):
        print("pay your insurance")
#child class
class Customer(Bank,Insurance):
    def pay(self):
        print("please pay your bank balance...!!!")
        
#create a object
c=Customer()

#call the inherits from child class
c.pay()
c.pay()
```

```python
#10.Demonstrate Method Resolution Order (MRO) using multiple inheritance with three classes.

class A:
    def show(self):
        print("Method from Class A")


class B(A):
    def show(self):
        print("Method from Class B")


class C(A):

    def show(self):
        print("Method from Class C")


class D(B, C):
    pass

# Create object
obj = D()

obj.show()

# Check MRO
print(D.mro())

```
