# Hierarchical inheritance

```python
#task1.Person → Student, Teacher
#Question
#Create a parent class Person with attributes name and age. Create two child classes Student and teacher that inherits from Person.
##Concept
#• Hierarchical Inheritance
##• Parent attributes inherited by multiple child classes.
#How to Solve
#1. Create a parent class Person.
#2. Add name and age using a constructor.
#3. Create Student(Person).
#4. Create Teacher(Person).
#5. Create objects of both classes.
#6. Show that both children can access name and age.


#parent class
class Person:
    #constructor
    def __init__(self,name,age):
        self.name=name
        self.age=age
        
#child class1
class Student(Person):
    pass
    
#child class2
class Teacher(Person):
    pass


#create student object
Student=Student("Shrutiii",21)

#create Teacher object
Teacher=Teacher("Vaishnaviiiiiiii",21)

#access inherits attributes
print("Student Name:",Student.name)
print("Student Age:",Student.age)

print("Teacher Name:",Teacher.name)
print("Teacher Age:",Teacher.age)
```

```python
#task2.Animal → Dog, Cat
#Question
#Define class Animal with method eat(). Create subclasses Dog and Cat.
#Concept
#• Method inheritance.
#How to Solve
#1. Create parent class Animal.
#2. Write method eat().
#3. Create child classes Dog and Cat.
#4. Create objects.
#5. Call eat() using both objects.


#parent class
class Animal:
    def eat(self):
        print("animal is eating")
#child class
class Dog(Animal):
    pass

#child class
class Cat(Animal):
    pass
    
#create a object
dog=Dog()
cat=Cat()
#display all information
dog.eat()
cat.eat()       
```

# Hierarchical Inheritance with Constructors & Methods

```python
#13Task.Person Constructor
#Concept
#Using super() to call the parent constructor.
#How to Solve
#1. Create constructor in parent.
#2. Create child constructors.
#3. Use super().__init__().
#4. Add child-specific attribute.
#5. Display all data.


#parent class                                              
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_person(self):
        print("Name:", self.name)
        print("Age:", self.age)


# Child class
class Student(Person):

    def __init__(self, name, age, course):
        # Call parent constructor
        super().__init__(name, age)

        # Child-specific attribute
        self.course = course

    def display_student(self):
        self.display_person()
        print("Course:", self.course)


# Create object
s1 = Student("Vaishnavi", 21, "Data Science")

# Display all data
s1.display_student()
```

```python
#14.Animal Sound Override
#Concept
#Method Overriding.
#How to Solve
#1. Parent has sound().
#2. Dog overrides.
#3. Cat overrides.
#4. Call method from each object.


#parent class
class Animal:
    def sound(self):
        print("animal is breaking")

#child class
class Dog(Animal):
    def sound(self):
        print(" Dog says::Wooofffff...!!!!!!!!!!")

#child class
class Cat(Animal):
    def sound(self):
        print(" Cat says::Memooooo.....!!!")

#create the object
d=Dog()
c=Cat()

#display all information
d.sound()
c.sound()
        
```

```python
#15.Vehicle Constructor
#Concept
#Constructor inheritance.
#How to Solve
#1. Parent stores brand.
#2. Car adds doors.
#3. Bike adds engine_cc.
#4. Use super().


#Parent class
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

# Child class - Car1
class Car(Vehicle):

    def __init__(self, brand, doors):
        super().__init__(brand)
        self.doors = doors

    def display_car(self):
        print("Brand:", self.brand)
        print("Doors:", self.doors)
            
# child class - Bike2
class Bike(Vehicle):

    def __init__(self, brand, engine_cc):
        super().__init__(brand)
        self.engine_cc = engine_cc

    def display_bike(self):
        print("Brand:", self.brand)
        print("Engine:", self.engine_cc, "cc")

#Create objects
car = Car("Toyota", 4)
bike = Bike("Honda", 150)


# Display data
car.display_car()

print()

bike.display_bike()
```
