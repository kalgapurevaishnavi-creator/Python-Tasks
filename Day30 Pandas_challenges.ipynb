```python
#20.Challenge 20 – Final Interview Project
#Your manager gives you a dirty customer dataset.
#Prepare the final dataset.
#Requirements
#Load dataset
#Check information
#Find missing values
#Remove duplicates
#Rename
#CustomerID
#AnnualIncome
#CreditScore
#Target
#Convert
#Age

#AnnualIncome
#Balance

#Fill missing values
#Remove
#Segment
#CustomerID
#Create new columns
#TotalMoney = Balance + AnnualIncome
#AgeAfter5Years = Age + 5
#MonthlyIncome = AnnualIncome / 12
#CreditScoreDouble = CreditScore * 2Display
#• Head
#• Tail
#• Sample
#• Describe
#• Target counts
#• Final information


import pandas as pd
import numpy as np

# 1. Load dataset
df = pd.read_csv(r"C:\Users\Dell\Downloads\Task_dataset.csv")

# Extra spaces from column names
df.columns = df.columns.str.strip()

# 2. Check information
print("Initial Information:")
df.info()

# 3. Find missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 4. Remove duplicates
df = df.drop_duplicates()

# 5. Rename columns
df = df.rename(columns={
    "CustomerID": "Customer_ID",
    "AnnualIncome": "Annual_Income",
    "CreditScore": "Credit_Score",
    "Target": "Output"
})

# 6. Handle infinite values
df = df.replace([np.inf, -np.inf], np.nan)

# 7. Fill missing values
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Annual_Income"] = df["Annual_Income"].fillna(df["Annual_Income"].median())
df["Balance"] = df["Balance"].fillna(df["Balance"].mean())

# 8. Convert data types
df["Age"] = df["Age"].astype(int)
df["Annual_Income"] = df["Annual_Income"].astype(float)
df["Balance"] = df["Balance"].astype(float)

# 9. Remove Segment and CustomerID
df = df.drop(columns=["Segment", "Customer_ID"])

# 10. Create new columns
df["TotalMoney"] = df["Balance"] + df["Annual_Income"]
df["AgeAfter5Years"] = df["Age"] + 5
df["MonthlyIncome"] = df["Annual_Income"] / 12
df["CreditScoreDouble"] = df["Credit_Score"] * 2

# 11. Head
df.head()

# 12. Tail
df.tail()

# 13. Sample
df.sample(5)

# 14. Describe
df.describe()

# 15. Target counts
df["Output"].value_counts()

# 16. Final Information
df.info()
```
