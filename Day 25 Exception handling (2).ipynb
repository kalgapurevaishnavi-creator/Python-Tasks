# Raising Exceptions & Custom Exceptions (19-25)

```python
## 19.Write a function that takes age and raises ValueError if age is less than 18.

def check_age(age):
    if age < 18:
        raise ValueError("18  above.")
try:        
    age = int(input("Enter your age: "))
    print(check_age(age))
    
except ValueError:
    print("ValueError")
```

```python
#20.Create a custom exception called NegativeNumberError and raise it when user enters a
#negative number.

class NegativeNumberError(Exception):
    pass
try:
    num = int(input("Enter a number: "))
    if num < 0:
        raise NegativeNumberError("Negative not allowed.")
    print("Number:", num)

except NegativeNumberError as e:
    print(e)
```

```python
#21Creat.e a custom exception InsufficientBalanceError for a banking program.

class InsufficientBalanceError(Exception):
    pass

try:
    balance = 500
    withdraw = int(input("Enter withdrawal amount: "))

    if withdraw > balance:
        raise InsufficientBalanceError("Insufficient balance.")

    balance = balance - withdraw
    print("Remaining Balance:", balance)

except InsufficientBalanceError as e:
    print(e)
```

```python
#22.Write a program that raises TypeError manually if the input is not a string.

try:
    data = 123

    if not isinstance(data, str):
        raise TypeError("Input must be a string.")
    print(data)

except TypeError as e:
    print("Error:",e)
```

```python
#23.Create a function that checks password length and raises a custom exception if length is
#less than 8.

class PasswordLengthError(Exception):
    pass
def check_password(password):
    if len(password) < 5:
        raise PasswordLengthError("Password must be at least 5 characters.")
    return "Password is valid."

try:
    password = input("Enter password: ")
    print(check_password(password))

except PasswordLengthError as e:
    print("Error:",e)
```
