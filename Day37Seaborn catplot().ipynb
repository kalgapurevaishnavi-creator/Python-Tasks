 # Seaborn catplot()

```python
#A.Titanic Dataset — 15 Questions
#Use:
import seaborn as sns
titanic = sns.load_dataset("titanic")
titanic
```

```python
#1.Survival by Gender
#Use catplot() to show the number of passengers who survived and did not survive for each
#gender.
#• X-axis → sex
#• Hue → survived
#• Chart → count

import seaborn as sns
titanic = sns.load_dataset("titanic")


import matplotlib.pyplot as plt
import seaborn as sns
sns.catplot(
    data=titanic,
    x="sex",
    hue="survived",
    kind="count"
)

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
#2.Survival by Passenger Class
#Create a categorical plot showing the number of passengers in each class.
#Use:
#• x="class"
#• kind="count"

import seaborn as sns
import matplotlib.pyplot as plt
df = sns.load_dataset("titanic")

sns.catplot(
    data=df,
    x="class",
    kind="count"
)

plt.title("Passenger Count by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
#3.Survival Count by Class
#Show the number of survivors and non-survivors for each passenger class.
#Use:
#x="class"
#hue="survived"
#kind="count"


sns.catplot(
    data=df,
    x="class",
    hue="survived",
    kind="count"
)

plt.title("Survival Count by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
##### 4.Survival Percentage by Gender
#Create a categorical visualization comparing survival across sex.
#Use catplot() and choose an appropriate categorical plot.
#Then identify which gender has a higher survival rate.


import seaborn as sns
import matplotlib.pyplot as plt

# Calculate survival percentage
survival_rate = df.groupby("sex")["survived"].mean().reset_index()

# Convert to percentage
survival_rate["survived"] = survival_rate["survived"] * 100

# Categorical plot
sns.catplot(
    data=survival_rate,
    x="sex",
    y="survived",
    kind="bar"
)

plt.title("Survival Percentage by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Percentage (%)")
plt.show()
```

```python
####5.Age Distribution by Gender
#Use catplot() to compare passenger age across different genders.
#Try an appropriate kind for showing the distribution.

sns.catplot(
    data=df,
    x="sex",
    y="age",
    kind="box"
)

plt.title("Age Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Age")
plt.show()
```

```python
#6.Age Distribution by Passenger Class
#Create a categorical plot showing age distribution for:
#• First class
#• Second class
#• Third class


import seaborn as sns
import matplotlib.pyplot as plt

sns.catplot(
    data=titanic,
    x='class',
    y='age',
    kind='box',
    height=6,
    aspect=1.2
)

plt.title('Age Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.show()
```

```python
#7.Fare by Passenger Class
#Use catplot() to compare fare across the three passenger classes.
#Add:
#• Appropriate color
#• Title
#• X/Y labels

import seaborn as sns
import matplotlib.pyplot as plt

sns.catplot(
    data=titanic,
    x='class',
    y='fare',
    kind='box',
    color='skyblue',
    height=6,
    aspect=1.2
)

plt.title('Fare Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.show()
```

```python
#9.Age vs Passenger Class
#Use a categorical plot to compare the distribution of age across class.
#Add sex using hue.

import matplotlib.pyplot as plt
import seaborn as sns

sns.catplot(
    data=titanic,
    x='class',
    y='age',
    hue='sex',
    kind='box',
    height=6,
    aspect=1.2
)

plt.title('Age Distribution by Passenger Class and Gender')
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.show()
```

```python
#10.Survival by Class and Gender
#Create a categorical visualization showing survival for each passenger class and gender.
#Use:
#x="class"
#hue="sex"
#Use an appropriate kind.


sns.catplot(
    data=titanic,
    x="class",
    hue="sex",
    col="survived",
    kind="count",
    height=5,
    aspect=1
)

plt.suptitle("Survival by Passenger Class and Gender", y=1.05)
plt.xlabel("Passenger Class")
plt.ylabel("Passenger Count")
plt.show()
```

```python
#13.Survival vs Age Group
#Create an age_group column:
#• Child → Age < 18
#• Adult → 18–59
#• Senior → 60+
#Use catplot() to compare survival across these groups.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create age groups
titanic["age_group"] = pd.cut(
    titanic["age"],
    bins=[0, 17, 59, float("inf")],
    labels=["Child", "Adult", "Senior"]
)

# Compare survival across age groups
sns.catplot(
    data=titanic,
    x="age_group",
    hue="survived",
    kind="count",
    height=6,
    aspect=1.2
)

plt.title("Survival by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Passenger Count")
plt.show()
```

```python
#14.Class, Gender and Survival
#Create a categorical visualization where:
#• X → class
#• Y → survival-related measure
#• Hue → sex
#• Columns → survived
#Use col= to create separate panels.

#create a categorical visulatization
sns.catplot(
    data=titanic,
    x='class',
    y='age',                 # survival-related measure; can also use another numeric column
    hue='sex',
    col='survived',
    kind='box',
    height=5,
    aspect=1
)

plt.show()
```

```python
#15.Titanic Multi-Panel Analysis
#Create a catplot() with:
#• X → class
#• Y → fare
#• Hue → sex
#• Columns → survived
#Use an appropriate categorical kind.
#Add:
#• Figure size
#• Title
#• Color palette
#• Appropriate labels
#Explain what you observe.

import seaborn as sns
import matplotlib.pyplot as plt

g = sns.catplot(
    data=titanic,
    x='class',
    y='fare',
    hue='sex',
    col='survived',
    kind='box',
    palette='Set2',
    height=5,
    aspect=1.2
)

g.fig.suptitle('Titanic Fare by Class, Gender and Survival', y=1.05)

g.set_axis_labels('Passenger Class', 'Fare')

g.set_titles('Survived = {col_name}')

plt.show()
```
