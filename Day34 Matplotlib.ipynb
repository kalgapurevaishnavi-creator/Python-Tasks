```python
#14.Create a graph containing three mathematical functions:
#Use different markers/line styles and add a legend.

import numpy as np
import matplotlib.pyplot as plt

# Values of x
x = np.arange(-10, 11)

# Three mathematical functions
y1 = x ** 2
y2 = x ** 3
y3 = 2 * x + 1

# Plot three functions
plt.plot(x, y1, marker='o', linestyle='-', label='y = x²')
plt.plot(x, y2, marker='s', linestyle='--', label='y = x³')
plt.plot(x, y3, marker='^', linestyle=':', label='y = 2x + 1')

# Add title and labels
plt.title("Three Mathematical Functions")
plt.xlabel("x")
plt.ylabel("y")

# Add legend
plt.legend()

# Add grid
plt.grid()
plt.show()    
```

```python
#10.Create a 2×2 subplot containing:
#o Line chart
#o Bar chart
#o Scatter plot
#o Histogram

import numpy as np
import matplotlib.pyplot as plt

# Data
x = np.arange(1, 7)
y_line = [10, 15, 13, 18, 20, 25]

categories = ["A", "B", "C", "D", "E"]
values = [20, 35, 25, 40, 30]

x_scatter = [1, 2, 3, 4, 5, 6]
y_scatter = [12, 18, 15, 22, 20, 28]

random_values = np.random.randn(100)

# Create 2x2 subplot
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# 1. Line chart
axes[0, 0].plot(x, y_line, marker="o")
axes[0, 0].set_title("Line Chart")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Sales")
axes[0, 0].grid(True)

# 2. Bar chart
axes[0, 1].bar(categories, values)
axes[0, 1].set_title("Bar Chart")
axes[0, 1].set_xlabel("Category")
axes[0, 1].set_ylabel("Value")

# 3. Scatter plot
axes[1, 0].scatter(x_scatter, y_scatter, s=80)
axes[1, 0].set_title("Scatter Plot")
axes[1, 0].set_xlabel("X")
axes[1, 0].set_ylabel("Y")
axes[1, 0].grid(True)

# 4. Histogram
axes[1, 1].hist(random_values, bins=10, edgecolor="black")
axes[1, 1].set_title("Histogram")
axes[1, 1].set_xlabel("Value")
axes[1, 1].set_ylabel("Frequency")

# Adjust spacing
plt.tight_layout()

# Display
plt.show()
```

```python
#11.Create a monthly sales chart and identify the maximum-sales month.
#Highlight that point using annotate().


import matplotlib.pyplot as plt

# Monthly sales data
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

sales = [120, 150, 140, 180, 200, 220,
         210, 250, 270, 260, 300, 320]

# Find maximum sales
max_sales = max(sales)
max_index = sales.index(max_sales)
max_month = months[max_index]

# Create line chart
plt.figure(figsize=(12, 6))

plt.plot(
    months,
    sales,
    marker="o",
    linewidth=2,
    label="Monthly Sales"
)

# Highlight maximum-sales point
plt.scatter(
    max_month,
    max_sales,
    s=100,
    zorder=5
)

# Annotate maximum sales
plt.annotate(
    f"Maximum Sales\n{max_month}: {max_sales}",
    xy=(max_month, max_sales),
    xytext=(-70, -50),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->")
)

# Title and labels
plt.title("Monthly Sales Performance")
plt.xlabel("Month")
plt.ylabel("Sales")

# Grid and legend
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()

plt.show()

# Print maximum-sales month
print("Maximum Sales Month:", max_month)
print("Maximum Sales:", max_sales)
```

```python
#12.Create a scatter plot of Age vs Income. Highlight customers whose income is
#greater than ₹1,00,000 using a different marker.


import matplotlib.pyplot as plt

# Customer data
age = [22, 25, 28, 32, 35, 38, 42, 45, 48, 52]
income = [40000, 55000, 70000, 85000, 95000,
          110000, 125000, 140000, 160000, 180000]

# Separate customers based on income
normal = [i <= 100000 for i in income]
high_income = [i > 100000 for i in income]

# Create scatter plot
plt.figure(figsize=(10, 6))

# Income <= ₹1,00,000
plt.scatter(
    [age[i] for i in range(len(age)) if normal[i]],
    [income[i] for i in range(len(income)) if normal[i]],
    marker="o",
    label="Income ≤ ₹1,00,000"
)

# Income > ₹1,00,000
plt.scatter(
    [age[i] for i in range(len(age)) if high_income[i]],
    [income[i] for i in range(len(income)) if high_income[i]],
    marker="*",
    s=150,
    label="Income > ₹1,00,000"
)

# Title and labels
plt.title("Age vs Income")
plt.xlabel("Age")
plt.ylabel("Income (₹)")

# Grid and legend
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()
```

```python
#13.Create a stacked bar chart showing monthly sales from:
#o Online
#o Offline
#o Corporate


import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

online = [12000, 15000, 14000, 18000, 20000, 22000]
offline = [8000, 10000, 9000, 12000, 13000, 15000]
corporate = [5000, 7000, 6000, 8000, 9000, 11000]

# Create stacked bar chart
plt.bar(months, online, label="Online")
plt.bar(months, offline, bottom=online, label="Offline")

# Calculate bottom for Corporate
bottom_corporate = [online[i] + offline[i] for i in range(len(months))]
plt.bar(months, corporate, bottom=bottom_corporate, label="Corporate")

plt.title("Monthly Sales by Channel")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
#14.Create a graph containing actual sales and target sales for 12 months.
#Highlight the months where actual sales are below the target.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

actual_sales = [12000, 15000, 14000, 19000, 18000, 22000,
                21000, 23000, 25000, 24000, 28000, 30000]

target_sales = [13000, 14000, 15000, 18000, 20000, 21000,
                22000, 24000, 23000, 26000, 27000, 29000]

# Plot actual and target sales
plt.figure(figsize=(10, 6))

plt.plot(months, actual_sales, marker="o", label="Actual Sales")
plt.plot(months, target_sales, marker="s", linestyle="--",
         label="Target Sales")

# Highlight months where actual sales are below target
for i in range(len(months)):
    if actual_sales[i] < target_sales[i]:
        plt.scatter(months[i], actual_sales[i],
                    marker="x", s=100, label="Below Target" if i == 0 else "")

plt.title("Actual Sales vs Target Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)

plt.show()
```

```python
#15.Create a mini analytics dashboard using Matplotlib with 4 subplots:
#o Monthly revenue line chart
#o Category-wise sales bar chart
#o Customer-age histogram
#o Revenue vs discount scatter plot

#Add titles, labels, legends, grids, and appropriate figure sizing to make the
#dashboard readable.



import matplotlib.pyplot as plt

# -----------------------------
# Sample Data
# -----------------------------

# 1. Monthly Revenue
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

revenue = [12000, 15000, 14000, 18000, 20000, 22000,
           21000, 25000, 23000, 27000, 30000, 32000]

# 2. Category-wise Sales
categories = ["Electronics", "Clothing", "Furniture", "Grocery"]
sales = [45000, 32000, 28000, 38000]

# 3. Customer Age
age = [22, 25, 28, 30, 32, 35, 38, 40, 42, 45,
       27, 29, 31, 36, 39, 41, 44, 26, 33, 37]

# 4. Revenue vs Discount
discount = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]
revenue_discount = [30000, 29000, 28000, 27000, 26000,
                    25000, 24000, 23000, 22000, 21000]


# -----------------------------
# Create Dashboard
# -----------------------------

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Monthly Revenue Line Chart
axes[0, 0].plot(months, revenue, marker="o", label="Revenue")
axes[0, 0].set_title("Monthly Revenue")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Revenue")
axes[0, 0].legend()
axes[0, 0].grid(True)

# 2. Category-wise Sales Bar Chart
axes[0, 1].bar(categories, sales, label="Sales")
axes[0, 1].set_title("Category-wise Sales")
axes[0, 1].set_xlabel("Category")
axes[0, 1].set_ylabel("Sales")
axes[0, 1].legend()
axes[0, 1].grid(axis="y")

# 3. Customer Age Histogram
axes[1, 0].hist(age, bins=6, label="Customers")
axes[1, 0].set_title("Customer Age Distribution")
axes[1, 0].set_xlabel("Age")
axes[1, 0].set_ylabel("Number of Customers")
axes[1, 0].legend()
axes[1, 0].grid(axis="y")

# 4. Revenue vs Discount Scatter Plot
axes[1, 1].scatter(discount, revenue_discount, label="Customers")
axes[1, 1].set_title("Revenue vs Discount")
axes[1, 1].set_xlabel("Discount (%)")
axes[1, 1].set_ylabel("Revenue")
axes[1, 1].legend()
axes[1, 1].grid(True)

# Main dashboard title
fig.suptitle("Mini Analytics Dashboard", fontsize=18)

# Adjust spacing
plt.tight_layout()
plt.show()
```
