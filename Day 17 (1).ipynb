```python
#24.In class BankAccount, add getter and setter for private __balance with validation.

class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    # Getter
    def get_balance(self):
        return self.__balance

    # Setter with validation
    def set_balance(self, balance):
        if balance >= 0:
            self.__balance = balance
        else:
            print("Balance cannot be negative.")


# Create object
account1 = BankAccount(1000)

#Get balance
print("Balance:", account1.get_balance())

# Update balance
account1.set_balance(1500)

print("Updated Balance:", account1.get_balance())

# Invalid balance
account1.set_balance(-500)
```

```python
#26.Define class Temperature with private __celsius. Add getter and setter with conversion
class Temperature:

    def __init__(self, celsius):
        self.__celsius = celsius

    # Getter
    def get_celsius(self):
        return self.__celsius

    # Setter
    def set_celsius(self, celsius):
        self.__celsius = celsius

    # Convert Celsius to Fahrenheit
    def to_fahrenheit(self):
        return (self.__celsius * 9 / 5) + 32

# Create object
temp1 = Temperature(25)

print("Celsius:", temp1.get_celsius())
print("Fahrenheit:", temp1.to_fahrenheit())

# Update temperature
temp1.set_celsius(30)

print("Updated Celsius:", temp1.get_celsius())
print("Updated Fahrenheit:", temp1.to_fahrenheit())
```

```python
#27.Create class Person with private __age. Add setter with validation (age between 0-120)

class Person:

    def __init__(self, age):
        self.__age = age

    # Setter with validation
    def set_age(self, age):
        if 0 <= age <= 120:
            self.__age = age
        else:
            print("Age must be between 0 and 120.")

    # Getter
    def get_age(self):
        return self.__age

# Create object
person1 = Person(22)

print("Age:", person1.get_age())

# Valid age
person1.set_age(25)
print("Updated Age:", person1.get_age())

# Invalid age
person1.set_age(150)
```

```python
#28.In class Product, add getter and setter for private __price with discount logic.

class Product:

    def __init__(self, price):
        self.__price = price

    # Getter
    def get_price(self):
        return self.__price

    # Setter with discount logic
    def set_price(self, price):
        if price > 500:
            self.__price = price * 0.90   # 10% discount
        else:
            self.__price = price
#Create object
product1 = Product(1000)

print("Price:", product1.get_price())

# Update price
product1.set_price(800)

print("Final Price:", product1.get_price())
```
