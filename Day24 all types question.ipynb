```python
#14.Create class BankAccount with private attribute __balance. Add deposit and withdraw
#methods with validation.


#class BankAccount
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # Private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposited:", amount)
        else:
            print("Invalid deposit amount!")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount!")
        elif amount > self.__balance:
            print("Insufficient balance!")
        else:
            self.__balance -= amount
            print("Withdrawn:", amount)
            
    def show_balance(self):
        print("Current Balance:", self.__balance)

# Create object
account = BankAccount(5000)

# Deposit money
account.deposit(2000)

# Withdraw money
account.withdraw(1000)

# Invalid withdrawal
account.withdraw(7000)

# Display balance
account.show_balance()
```

```python
#15.Create class Employee with private __salary. Add getter and setter. Setter should not
#allow salary below 5000.

class Employee:
    def __init__(self, salary):
        self.__salary = salary

    # Getter
    def get_salary(self):
        return self.__salary

    # Setter
    def set_salary(self, salary):
        if salary >= 5000:
            self.__salary = salary
            print("Salary updated successfully.")
        else:
            print("Salary cannot be less than 5000.")

# Create object
emp = Employee(10000)

# Getter
print("Current Salary:", emp.get_salary())

# Setter
emp.set_salary(4000)    # Invalid
emp.set_salary(15000)   # Valid

# Getter
print("Updated Salary:", emp.get_salary())
```

```python
#17.Create class Product with private __price. Add getter and setter with condition that price
#cannot be negative.

class Product:
    def __init__(self, price):
        self.__price = price

    # Getter
    def get_price(self):
        return self.__price

    # Setter
    def set_price(self, price):
        if price >= 0:
            self.__price = price
            print("Price updated successfully.")
        else:
            print("Price cannot be negative.")

# Create object
p1 = Product(50)

# Getter
print("Current Price:", p1.get_price())

# Setter
p1.set_price(-10)   # Invalid
p1.set_price(75)    # Valid

# Getter
print("Updated Price:", p1.get_price())
```

```python
#18.Create class Rectangle with private length and width. Add getter, setter and method to
#calculate area.
class Rectangle:
    def __init__(self, length, width):
        self.__length = length
        self.__width = width

    # Getter for length
    def get_length(self):
        return self.__length

    # Getter for width
    def get_width(self):
        return self.__width

    # Setter for length
    def set_length(self, length):
        if length > 0:
            self.__length = length
        else:
            print("Length must be greater than 0.")

    # Setter for width
    def set_width(self, width):
        if width > 0:
            self.__width = width
        else:
            print("Width must be greater than 0.")

    # Method to calculate area
    def calculate_area(self):
        return self.__length * self.__width


# Create object
r1 = Rectangle(10, 5)

# Getter
print("Length:", r1.get_length())
print("Width:", r1.get_width())

# Area
print("Area:", r1.calculate_area())

# Setter
r1.set_length(15)
r1.set_width(8)

# Updated Area
print("Updated Area:", r1.calculate_area())
```

# Inheritance + Abstraction + super() (19-24)

```python
#19.Create single level inheritance: Person → Student. Add attributes in both classes.

## Parent Class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Child Class
class Student(Person):
    def __init__(self, name, age, roll_no, course):
        super().__init__(name, age)
        self.roll_no = roll_no
        self.course = course

    def display(self):
        print("Name :", self.name)
        print("Age :", self.age)
        print("Roll No :", self.roll_no)
        print("Course :", self.course)

# Create object
s1 = Student("Vaishnavi", 22, 39, "MCA")

# Display details
s1.display()
```

```python
#20. Create multi-level inheritance: Person → Employee → Manager.

# Parent Class
class Person:
    def __init__(self, name):
        self.name = name

# Child Class
class Employee(Person):
    def __init__(self, name, emp_id):
        super().__init__(name)
        self.emp_id = emp_id

# Grandchild Class
class Manager(Employee):
    def __init__(self, name, emp_id, department):
        super().__init__(name, emp_id)
        self.department = department

    def display(self):
        print("Name:", self.name)
        print("Employee ID:", self.emp_id)
        print("Department:", self.department)

# Create object
m1 = Manager("Vaishnavi", 101, "HR")

# Display details
m1.display()
```

```python
#23.Create abstract class Shape with abstract method area(). Implement it in Circle and
#Rectangle using super().

from abc import ABC, abstractmethod
#Abstract Class
class Shape(ABC):
    def __init__(self):
        print("Shape Constructor")

    @abstractmethod
    def area(self):
        pass

# Child Class - Circle
class Circle(Shape):
    def __init__(self, radius):
        super().__init__()
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

# Child Class - Rectangle
class Rectangle(Shape):
    def __init__(self, length, width):
        super().__init__()
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

# Create objects
c = Circle(5)
print("Circle Area:", c.area())

r = Rectangle(10, 4)
print("Rectangle Area:", r.area())
```

```python
#24.Create hybrid inheritance with classes A, B, C and D (D inherits from B and C).

# Parent Class
class A:
    def showA(self):
        print("This is Class A")

# Child Class 1
class B(A):
    def showB(self):
        print("This is Class B")

# Child Class 2
class C(A):
    def showC(self):
        print("This is Class C")

# Hybrid Inheritance
class D(B, C):
    def showD(self):
        print("This is Class D")

# Create object
obj = D()

# Call methods
obj.showA()
obj.showB()
obj.showC()
obj.showD()
```

# Method Overloading, Overriding, Operator Overloading + Duck Typing (25-30)

```python
#25.Create a class Calculator that can add 2 or 3 numbers using method overloading concept
#(*args or default arguments).

class Calculator:

    def add(self, *args):
        if len(args) == 2:
            print("Sum of 2 numbers:", args[0] + args[1])
        elif len(args) == 3:
            print("Sum of 3 numbers:", args[0] + args[1] + args[2])
        else:
            print("Please enter only 2 or 3 numbers.")

# Create object
c = Calculator()

# Add 2 numbers
c.add(1,2)

# Add 3 numbers
c.add(1,2,3)

# Invalid input
c.add(1,6,7,8)
```

```python
#27.Create class Point and overload + operator to add two points.

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overload + operator
    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def display(self):
        print("Point(", self.x, ",", self.y, ")")

# Create objects
p1 = Point(2, 3)
p2 = Point(4, 5)

# Add two points
p3 = p1 + p2

# Display result
print("Point 1:")
p1.display()

print("Point 2:")
p2.display()

print("Result after Addition:")
p3.display()
```

```python
#28.Create class Book and overload > operator to compare number of pages of two books.

class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    # Overload > operator
    def __gt__(self, other):
        return self.pages > other.pages

# Create objects
book1 = Book("Python", 30)
book2 = Book("Java", 25)

# Compare books
if book1 > book2:
    print(book1.title, "has more pages than", book2.title)
else:
    print(book2.title, "has more pages than", book1.title)
```

```python
#30.Create class ComplexNumber and overload both + and - operators.

class ComplexNumber:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    # Overload + operator
    def __add__(self, other):
        return ComplexNumber(self.real + other.real,
                             self.imag + other.imag)

    # Overload - operator
    def __sub__(self, other):
        return ComplexNumber(self.real - other.real,
                             self.imag - other.imag)

    # Display complex number
    def display(self):
        print(f"{self.real} + {self.imag}i")


# Create objects
c1 = ComplexNumber(4, 5)
c2 = ComplexNumber(2, 3)

# Addition
add_result = c1 + c2

# Subtraction
sub_result = c1 - c2

# Display results
print("First Complex Number:")
c1.display()

print("Second Complex Number:")
c2.display()

print("After Addition:")
add_result.display()

print("After Subtraction:")
sub_result.display()
```
