```python
#Task1. Class Name: BankAccount 
#Problem 
#Create a class named BankAccount. 
#Explanation 
#A bank account stores money. The balance should not be directly accessible from outside  the class, so create a private attribute named __balance. 
#Implement the following methods: 
#• deposit(amount)  
#o Deposit money into the account.  
#o Amount must be greater than 0.  
#• withdraw(amount
#o Withdraw money.  
#o Withdrawal should happen only if:  
#▪ amount > 0  
#▪ amount <= available balance


class BankAccount:
    
    def __init__(self, ac_no, balance):
        self.__account = ac_no
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Successfully deposited {amount:.2f}")
        else:
            print("Deposit amount must be greater than 0.")

    def withdraw(self, amount):
        if amount > 0:
            if amount <= self.__balance:
                self.__balance -= amount
                print(f"Successfully withdrawn {amount:.2f}")
            else:
                print("Insufficient balance.")
        else:
            print("Withdrawal amount must be greater than 0.")

    def get_balance(self):
        return self.__balance


account = BankAccount(101, 0)

account.deposit(100)
account.withdraw(48)
account.withdraw(70)

print("Balance:", account.get_balance())
```

```python
#6.Class Name: Product 
#Problem 
#Create a class named Product. Explanation 
#Store 
#• private __price  
#• private __stock  
#Methods: 
#• update_price(new_price)  • update_stock(quantity)  
#Validation: 
#Price cannot be negative. 
#Stock cannot become negative. 

class Product:

    def __init__(self, price, stock):
        self.__price = price
        self.__stock = stock

    def update_price(self, new_price):
        if new_price >= 0:
            self.__price = new_price
            print("Price updated successfully.")
        else:
            print("Price cannot be negative.")

    def update_stock(self, quantity):
        if self.__stock + quantity >= 0:
            self.__stock += quantity
            print("Stock updated successfully.")
        else:
            print("Stock cannot become negative.")

    def display(self):
        print("Price:", self.__price)
        print("Stock:", self.__stock)


# Create object
product1 = Product(1000, 10)

product1.display()

# Update price
product1.update_price(1200)

# Update stock
product1.update_stock(-3)

product1.display()

```

```python
#16.. Class Name: Shape 
#Problem 
#Create an abstract class Shape. 
#Explanation 
#Declare abstract methods 
#• calculate_area()  
#• calculate_perimeter()  
#Create subclasses like 
#• Circle  
#• Rectangle  
#Implement both methods.


from abc import ABC, abstractmethod
import math
  
# Abstract class
class Shape(ABC):

    @abstractmethod
    def calculate_area(self):
        pass

    @abstractmethod
    def calculate_perimeter(self):
        pass


# Subclass: Circle
class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return math.pi * self.radius * self.radius

    def calculate_perimeter(self):
        return 2 * math.pi * self.radius


# Subclass: Rectangle
class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        return self.length * self.width

    def calculate_perimeter(self):
        return 2 * (self.length + self.width)


# Objects
circle = Circle(5)
rectangle = Rectangle(10, 5)

print("Circle Area:", circle.calculate_area())
print("Circle Perimeter:", circle.calculate_perimeter())

print("Rectangle Area:", rectangle.calculate_area())
print("Rectangle Perimeter:", rectangle.calculate_perimeter())
```

```python
#17. Class Name: Vehicle 
#Problem 
#Create abstract class Vehicle. 
#Explanation 
#Abstract methods 
#• start()  
#• stop()  
#Implement in 
#• Car  
#• Bike  
#Each subclass should define how the vehicle starts and stops. 


#Abstract class
class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass


# Subclass: Car
class Car(Vehicle):

    def start(self):
        print("Car starts with a key")

    def stop(self):
        print("Car stops using brakes")


# Subclass: Bike
class Bike(Vehicle):

    def start(self):
        print("Bike starts with a self-start button")

    def stop(self):
        print("Bike stops using brakes")


# Create objects
car = Car()
bike = Bike()

# Car methods
car.start()
car.stop()

# Bike methods
bike.start()
bike.stop()
```

```python
#18.Class Name: Payment 
#Problem 
#Create abstract class Payment. 
#Explanation 
#Abstract method 
#process_payment(amount) 
#Create subclasses 
#• CreditCardPayment  
#• UPIPayment  
#Each should process payment differently.
        
# Abstract class
class Payment(ABC):

    @abstractmethod
    def process_payment(self, amount):
        pass


# Credit Card Payment
class CreditCardPayment(Payment):

    def process_payment(self, amount):
        print(f"Processing credit card payment of ${amount}")


# UPI Payment
class UPI(Payment):

    def process_payment(self, amount):
        print(f"Processing UPI payment of ${amount}")


# Create objects
credit_card = CreditCardPayment()
upi = UPI()

#Call methods
credit_card.process_payment(700)
upi.process_payment(578)    
```

```python
#19.Class Name: Animal 
#Problem 
#Create abstract class Animal. Explanation 
#Abstract method 
#make_sound() 
#Create subclasses 
#• Dog  
#• Cat  
#Each returns its own sound.

from abc import ABC, abstractmethod
#Abstract class
class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        pass

# Dog class
class Dog(Animal):
    def make_sound(self):
        return "Dog says: Woof Woof"

# Cat class
class Cat(Animal):
    def make_sound(self):
        return "Cat says: Meow Meow"
        
# Create objects

dog = Dog()
cat = Cat()
print(dog.make_sound())
print(cat.make_sound())
```

```python
#20. Class Name: Report 
#Problem 
#Create abstract class Report. Explanation 
#Abstract method 
#generate_report() 
#Create 
#• PDFReport  
#• ExcelReport  
#Each generates report differently

#Abstract class
class Report(ABC):

    @abstractmethod
    def generate_report(self):
        pass


# Subclass: PDFReport
class PDFReport(Report):

    def generate_report(self):
        print("Generating PDF report")


# Subclass: ExcelReport
class ExcelReport(Report):

    def generate_report(self):
        print("Generating Excel report")


# Create objects
pdf = PDFReport()
excel = ExcelReport()

# Generate reports
pdf.generate_report()
excel.generate_report()
```
