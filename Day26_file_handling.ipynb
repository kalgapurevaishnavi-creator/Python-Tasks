```python
#5.Count the number of words in a text file.


# Count the number of words in a text file

# Open the file in read mode
file = open("sample.txt", "r")

# Read the content
text = file.read()

# Close the file
file.close()

# Split the text into words
words = text.split()

# Count the words
count = len(words)

# Display the result
print("Total number of words:", count)
```

```python
#9.Check if a file exists or not before reading it.

import os
filename = "details.txt"

if os.path.exists(filename):
    file = open(filename, "r")
    
    content = file.read()
    print(content)
    
    file.close()

else:
    print("File does not exist.")
```

```python
#16.Write a program to merge two text files into a third file.

#Create first file
file1 = open("file1.txt", "w")
file1.write("Hello Vaishnavi Kalgapure.\n")
file1.write("I am learning Python.")
file1.close()


# Create second file
file2 = open("file2.txt", "w")
file2.write("Python is used in Data Science.\n")
file2.write("File Handling is easy.")
file2.close()

print("Two files created successfully.")
```

```python
# Merge two text files into a third file
file1 = open("file1.txt", "r")
file2 = open("file2.txt", "r")

content1 = file1.read()
content2 = file2.read()

file1.close()
file2.close()

file3 = open("merged.txt", "w")

file3.write(content1)
file3.write("\n")
file3.write(content2)

file3.close()
print("Two files merged successfully.")
```

```python
#17.Store a list of numbers in a file, then read and calculate their sum.

numbers = [10, 20, 30, 40, 50]

# Write numbers into file
file = open("numbers.txt", "w")

for num in numbers:
    file.write(str(num) + "\n")

file.close()

# Read numbers from file and calculate sum
file = open("numbers.txt", "r")

total = 0

for line in file:
    total += int(line)

file.close()
print("Sum of numbers is:", total)
```

```python
#22.Write a program that reads a log file and counts the number of “ERROR” messages.

# Create log file
file = open("log.txt", "w")

file.write("INFO: Program started\n")
file.write("ERROR: File not found\n")
file.write("INFO: User logged in\n")
file.write("ERROR: Database connection failed\n")
file.write("ERROR: Invalid input")

file.close()
print("Log file created successfully.")


# Count number of ERROR messages in a log file
file = open("log.txt", "r")

count = 0

for line in file:
    if "ERROR" in line:
        count += 1

file.close()
print("Number of ERROR messages:", count)
```

```python
#23.Take multiple lines of text from user until they type “STOP”, then save all lines in a file.

file = open("text.txt", "w")

while True:
    line = input("Enter text: ")
    
    if line == "STOP":
        break
    
    file.write(line + "\n")

file.close()
print("All lines saved successfully.")
```

```python
#24.Write a program to encrypt the content of a file (simple character shifting) and save it in
#another file.

# Read original file
file = open("details.txt", "r")

content = file.read()

file.close()

# Encrypt by shifting each character by 3
encrypted = ""

for char in content:
    encrypted += chr(ord(char) + 3)

# Save encrypted content to another file
file = open("encrypted.txt", "w")

file.write(encrypted)

file.close()

print("File encrypted successfully.")
```

```python
#25.Read a file containing student marks, calculate average, and write the result in another
#file.

#create a file
file = open("marks.txt", "w")

file.write("vaishuu80\n")
file.write("shrutii:90\n")
file.write("jyotii:70\n")
file.write("aaruu:85")

file.close()

print("Marks file created successfully.")
```

```python
file = open("marks.txt", "r")

total = 0
count = 0

for line in file:
    name, marks = line.strip().split(":")
    total += int(marks)
    count += 1

file.close()
average = total / count

file = open("result.txt", "w")

file.write("Average Marks: " + str(average))
file.close()
print("Average calculated and saved successfully.")
```
