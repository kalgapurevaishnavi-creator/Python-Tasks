```python
##1.How many missing values are there in each column? Which column has the highest
##number of missing values?

import pandas as pd
df=pd.read_csv("C:\\Users\\Dell\\Downloads\\Customer_datasets.csv")

# Count missing values in each column
df.isnull().sum()
```

```python
#2.What percentage of rows have at least one missing value?

# Percentage of rows having at least one missing value
percentage = df.isnull().any(axis=1).mean() * 100
percentage
```

```python
#3.Drop all rows that have more than 3 missing values. How many rows remain?

# Drop rows having more than 3 missing values
df_clean = df.dropna(thresh=df.shape[1] - 3)

# Number of rows remaining
print("Rows remaining:", len(df_clean))
```

```python
##########Duplicates##############
```

```python
#4.How many exact duplicate rows are present in the dataset?

# Count exact duplicate rows
duplicates = df.duplicated().sum()
duplicates
```

```python
#5.How many duplicate Customer_IDs are there? Keep only the first occurrence of each
#Customer_ID.

# Count duplicate Customer_IDs
duplicate_ids = df["Customer_ID"].duplicated().sum()

print("Number of duplicate Customer_IDs:", duplicate_ids)

# Keep only the first occurrence of each Customer_ID
df = df.drop_duplicates(subset="Customer_ID", keep="first")

# Check rows remaining
print("Rows remaining:", len(df))
```

```python
###############Age Column###############################
```

```python
#6.Convert the Age column to numeric. How many values could not be converted
#(became NaN)?

# Convert Age column to numeric
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

# Count values that became NaN
invalid_age = df["Age"].isna().sum()
invalid_age
```

```python
#7.After converting Age to numeric, identify and remove outliers using the IQR method.
#How many outliers were removed?

# Convert Age to numeric
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

# Calculate Q1 and Q3
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Find lower and upper limits
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Count outliers
outliers = ((df["Age"] < lower) | (df["Age"] > upper)).sum()

print("Number of outliers:", outliers)

# Remove outliers
df = df[(df["Age"] >= lower) & (df["Age"] <= upper)]

print("Outliers removed:", outliers)
print("Rows remaining:", len(df))
```

```python
#8.Replace remaining missing values in Age with the median age.

# Replace missing Age values with median age
df["Age"] = df["Age"].fillna(df["Age"].median())
df
```

```python
#Salary Column
```

```python
#9.Clean the Salary column: remove currency symbols ($), commas, and convert
#everything possible to numeric. How many values remained non-numeric?

# Remove $ and commas, then convert to numeric
df["Salary"] = (
    df["Salary"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")

# Count values that remained non-numeric
non_numeric = df["Salary"].isna().sum()
non_numeric
```

```python
#10.After cleaning, find the mean and median salary. Which one is more affected by
#outliers?

# Calculate mean and median salary
mean_salary = df["Salary"].mean()
median_salary = df["Salary"].median()

mean_salary
median_salary
```

```python
#11.Convert the Join_Date column to proper datetime format. How many dates failed to
#parse?

# Convert Join_Date to datetime
df["Join_Date"] = pd.to_datetime(df["Join_Date"], errors="coerce")

# Count dates that failed to parse
failed_dates = df["Join_Date"].isna().sum()
failed_dates
```

```python
#12.Extract the year from the cleaned Join_Date. What is the most common joining year?

# Extract year from Join_Date
df["Join_Year"] = df["Join_Date"].dt.year

# Find the most common joining year
most_common_year = df["Join_Year"].mode()[0]
most_common_year
```

```python
#13.Standardize the Department column (fix casing, trim spaces, fix common typos like
#“Sale” → “Sales”, “Financ” → “Finance”, etc.). How many unique departments remain
#after cleaning?

import pandas as pd
df=pd.read_csv("C:\\Users\\Dell\\Downloads\\Customer_datasets.csv")


# Standardize Department column
df['Department'] = (
    df['Department']
    .astype(str)
    .str.strip()
    .str.title()
)

# Fix common typos
df['Department'] = df['Department'].replace({
    'Sale': 'Sales',
    'Sales ': 'Sales',
    'Financ': 'Finance',
    'Finace': 'Finance',
    'Hr': 'HR',
    'Human Resource': 'HR',
    'Human Resources': 'HR'
})

# Check unique departments
df['Department'].unique()

# Count unique departments
df['Department'].nunique()
```

```python
#14.Clean the City column (standardize names, fix “Los Angles” → “Los Angeles”, “NY” →
#“New York”, “Philly” → “Philadelphia”, etc.). List the final unique cities.


# Clean City column
df['City'] = df['City'].astype(str).str.strip().str.title()

# Fix common city names / typos
df['City'] = df['City'].replace({
    'Los Angles': 'Los Angeles',
    'Ny': 'New York',
    'Philly': 'Philadelphia',
    'Nyc': 'New York',
    'New York City': 'New York',
    'Los Angeles ': 'Los Angeles'
})

# Display final unique cities
df['City'].unique()
```

```python
#15.Standardize the Gender column so it only contains “Male”, “Female”, and “Other”.
#How many records were changed?

# Count original values
original_gender = df['Gender'].copy()

# Standardize Gender
df['Gender'] = df['Gender'].astype(str).str.strip().str.title()

# Fix common variations
df['Gender'] = df['Gender'].replace({
    'M': 'Male',
    'Man': 'Male',
    'F': 'Female',
    'Woman': 'Female',
    'O': 'Other',
    'Others': 'Other'
})

# Count how many records changed
changed_records = (original_gender != df['Gender']).sum()

print("Number of records changed:", changed_records)

# Check final values
df['Gender'].unique()
```
