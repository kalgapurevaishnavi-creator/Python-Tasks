```python
#Task1.Class Name: Person → Student
#Problem
#Create a parent class named Person and a child class named Student.
#Explanation
#The Person class should store:
#• name
#• age
#The Student class should inherit from Person and add:
#• marks
#Create a method in Student to display the student's name, age, and marks.

#parent class
class Person:
    #constructor
    def __init__(self,name,age):
        self.Name=name
        self.Age=age
        
#child class
class student(Person):
    #constructor
    def __init__(self,name,age,marks):
        super().__init__(name,age)
        self.stu_marks= marks
        
    def display_all_info(self):
        print(f"Name: {self.Name}")
        print(f"Age: {self.Age}")
        print(f"Marks: {self.stu_marks}")
        
#create a object
student = student("Vaishu",22,100)
student.display_all_info()
```

```python
#5.Class Name: Shape → Circle
#Problem
#Create a parent class named Shape.
#Explanation
#Create a method:
#• area()
#Create child class:
#• Circle
#Store:
#• radius
#Override area() to calculate the area of a circle.

import math
#parent class
class Shape:
    def area(self):
        print("shape is aera")
        
#child class
class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius
        
#method overloding
    def aera(self):
       result = math.pi * self.radius ** 2
       print(f"Area of circle:{result}")
    
#create a object
Circle1=Circle(11)

#call overridden method
Circle1.aera()
```

```python
#6.Class Name: BankAccount → SavingsAccount
#Problem
#Create a parent class named BankAccount.
#Explanation
#Provide methods:
#• deposit()
#• withdraw()
#Create subclass:
#• SavingsAccount
#Add:
#• calculate_interest()

#parent class
class Bankaccount:
    def __init__(self,balance):
        self.balance=balance
        
    def deposite(self,amount):
        self.balance =+ amount
        print("Deposite:",amount)
        print("Current balance:",self.balance)
        
    def withdraw(self,amount):
        if amount <= self.balance:
            self.balance -= amount
            print("withdrawn:",amount)
            print("Current balance:",self.balance)
        else:
            print("Insuffisiant balance")
            
#child class
class SavingAccount(Bankaccount):
     def calculate_intrest(self,rate):
        intrest=self.balance * rate/100
        print("Intrest:",intrest)
        
#class object
account= SavingAccount(110000)

#call the parent class method
account.deposite(60000)
account.withdraw(300)
#calculate child class method
account.calculate_intrest(2)
```

```python
#9.Class Name: Device → Mobile Problem 
#Create a parent class named Device.
#Explanation 
#Create method: 
#• turn_on()  
#Create subclass: 
#• Mobile  
#Add method: 
#• call(number).

# Parent class
class Device:
    def __init__(self, name):
        self.name = name

    def turn_on(self):
        print(self.name, "is turned on")

# Subclass
class Mobile(Device):
    def __init__(self, name, number):
        super().__init__(name)
        self.number = number

    def call(self):
        print("Calling number:", self.number)

    def display_all_info(self):
        print("Device Name:", self.name)
        print("Mobile Number:", self.number)
        
# Create object
m1 = Mobile("Vivo", 8421444)

#Call methods
m1.turn_on()
m1.call()
```

```python
#12.Class Name: Account → LoanAccount Problem 
#Create a parent class named Account. Explanation 
#Store: 
#• balance  
#Create subclass:
#• LoanAccount  
#Add: 
#• calculate_emi()

#Parent class
class Account:
    def __init__(self, balance):
        self.balance = balance

# Subclass
class LoanAccount(Account):
    def __init__(self, balance, calculate_emi):
        super().__init__(balance)   # Parent constructor call

        self.cal = calculate_emi

    def display(self):
        print("Balance:", self.balance)
        print("EMI:", self.cal)

# Class object
l1 = LoanAccount(4007, 500)
# Display
l1.display()
```

```python
#13.. Class Name: Employee → Developer Problem 
#Create a parent class named Employee. Explanation 
#Create method: 
#• get_salary()  
#Create subclass: 
#• Developer  
#Add: 
#• bonus  
#Display total salary including bonus. 

# Parent class
class Employee:
    def __init__(self, salary):
        self.salary = salary

    def get_salary(self):
        return self.salary

# Subclass
class Developer(Employee):
    def __init__(self, salary, bonus):
        super().__init__(salary)   # Parent constructor call
        self.bonus = bonus

    def total_salary(self):
        total = self.get_salary() + self.bonus
        print("Total Salary:", total)
# Object
d1 = Developer(50000, 10000)

# Display total salary
d1.total_salary()
```

```python
#Task 16.Class Name: 
#Person → Employee → Manager 
#Problem 
#Create a three-level inheritance hierarchy. Explanation 
#Person 
#Store: 
#• name  
#↓ 
#Employee 
#Add:
#• employee_id  
#↓ 
#Manager 
#Add: 
#• department  
#Display all information

# Grandparent class
class Person:
    def __init__(self, name):
        self.name = name

# Parent class
class Employee(Person):
    def __init__(self, name, employee_id):
        super().__init__(name)
        self.employee_id = employee_id

# Child class
class Manager(Employee):
    def __init__(self, name, employee_id, department):
        super().__init__(name, employee_id)
        self.department = department

    def display(self):
        print("Name:", self.name)
        print("Employee ID:", self.employee_id)
        print("Department:", self.department)

# Object
m1 = Manager("Vaishnavi", 101, "IT")
# Display all information
m1.display()
```

```python
#17.Class Name: 
#Animal → Mammal → Dog 
#Problem 
#Create multi-level inheritance. 
#Explanation 
#Animal 
#• eat()  
#↓ 
#Mammal 
#• walk()  
#↓ 
#Dog 
#• bark()  
#Demonstrate all inherited methods. 

# Grandparent class
class Animal:
    def eat(self):
        print("Animal is eating")
# Parent class
class Mammal(Animal):
    def walk(self):
        print("Mammal is walking")

# Child class
class Dog(Mammal):
    def bark(self):
        print("Dog is barking")
# Object
d1 = Dog()

# Demonstrate all inherited methods
d1.eat()
d1.walk()
d1.bark()
```
