```python
#11.The Age column contains numbers as well as values like "unknown" and
#"twenty five". Clean this column and convert it into a proper numeric column.

import pandas as pd

# Create a copy of the DataFrame
df = df.copy()

# Convert Age to numeric
df["age"] = pd.to_numeric(df["age"], errors="coerce")

# Display the cleaned Age column
df["age"].head()

# Check data type
df["age"].dtype
```

```python
#12.The Annual Income column contains values such as 650000 INR, ₹720000,
#1.2L, and N/A. Clean the column and convert the valid values into numeric
#format.

import pandas as pd
import numpy as np

def clean_income(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    # Missing values
    if value in ["n/a", "na", "nan", "unknown", ""]:
        return np.nan

    # Remove INR and  symbols
    value = value.replace("inr", "")
    value = value.replace("₹", "")
    value = value.replace(",", "").strip()

    # Convert Lakh values
    if value.endswith("l"):
        return float(value[:-1]) * 100000

    # Convert normal numeric values
    return pd.to_numeric(value, errors="coerce")


# Clean Annual Income
df["annual_income"] = df["annual_income"].apply(clean_income)

# Check results
print(df["annual_income"].head(10))

# Check data type
print("Data type:", df["annual_income"].dtype)
```

```python
#27.Detect outliers in Annual Income using the IQR method.

import pandas as pd

# Convert Annual Income to numeric
df['Annual Income'] = pd.to_numeric(
    df['Annual Income'],
    errors='coerce'
)

# Calculate Q1 and Q3
Q1 = df['Annual Income'].quantile(0.25)
Q3 = df['Annual Income'].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate lower and upper limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Detect outliers
outliers = df[
    (df['Annual Income'] < lower_limit) |
    (df['Annual Income'] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

print("\nOutliers in Annual Income:")
print(outliers)
```

```python
#28.Detect outliers in Loan Amount using the IQR method.

import pandas as pd

# Convert Loan Amount to numeric
df['Loan Amount (INR)'] = pd.to_numeric(
    df['Loan Amount (INR)'],
    errors='coerce'
)

# Calculate Q1 and Q3
Q1 = df['Loan Amount (INR)'].quantile(0.25)
Q3 = df['Loan Amount (INR)'].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Detect outliers
outliers = df[
    (df['Loan Amount (INR)'] < lower_limit) |
    (df['Loan Amount (INR)'] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

print("\nOutliers in Loan Amount:")
print(outliers)
```

```python
#30.Final cleaning task: Create a completely cleaned version of the dataset by:
#• Cleaning column names
#• Removing duplicates
#• Handling missing values
#• Fixing data types
#• Cleaning numerical columns
#• Standardizing categorical values
#• Handling invalid values
#• Detecting and treating outliers
#• Checking the final dataset for remaining errors

import pandas as pd

df = pd.read_csv(r"C:\Users\Dell\Downloads\Task_dataset.csv")

print(df.head())

# ============================================
# 1. LOAD DATASET
# ============================================

# Use your actual file name
df = pd.read_csv("your_file.csv")

# ============================================
# 2. CLEAN COLUMN NAMES
# ============================================

df.columns = (
    df.columns
    .str.strip()
    .str.replace(r'\s+', '_', regex=True)
    .str.replace(r'[^\w_]', '', regex=True)
)

print("Cleaned columns:")
print(df.columns.tolist())


# ============================================
# 3. REMOVE DUPLICATES
# ============================================

duplicates = df.duplicated().sum()
print("Duplicate rows:", duplicates)

df = df.drop_duplicates()


# ============================================
# 4. CLEAN NUMERICAL COLUMNS
# ============================================

# Age
df['Age'] = pd.to_numeric(
    df['Age'],
    errors='coerce'
)

# Annual Income
df['Annual_Income'] = (
    df['Annual_Income']
    .astype(str)
    .str.replace(r'[^0-9.]', '', regex=True)
)

df['Annual_Income'] = pd.to_numeric(
    df['Annual_Income'],
    errors='coerce'
)

# Credit Score
df['credit_Score'] = pd.to_numeric(
    df['credit_Score'],
    errors='coerce'
)

# Loan Amount
df['Loan_Amount_INR'] = (
    df['Loan_Amount_INR']
    .astype(str)
    .str.replace(r'[^0-9.]', '', regex=True)
)

df['Loan_Amount_INR'] = pd.to_numeric(
    df['Loan_Amount_INR'],
    errors='coerce'
)

# Monthly Visits
df['Monthly_Visits'] = df['Monthly_Visits'].replace({
    'one': 1,
    'two': 2,
    'three': 3,
    'four': 4,
    'five': 5
})

df['Monthly_Visits'] = pd.to_numeric(
    df['Monthly_Visits'],
    errors='coerce'
)


# ============================================
# 5. CLEAN DEBT TO INCOME
# ============================================

df['debt_to_income'] = (
    df['debt_to_income']
    .astype(str)
    .str.strip()
    .str.replace('%', '', regex=False)
    .str.replace('x', '', regex=False)
)

df['debt_to_income'] = pd.to_numeric(
    df['debt_to_income'],
    errors='coerce'
)

# Convert percentage values to decimal
df.loc[
    df['debt_to_income'] > 1,
    'debt_to_income'
] = df.loc[
    df['debt_to_income'] > 1,
    'debt_to_income'
] / 100


# ============================================
# 6. STANDARDIZE CATEGORICAL COLUMNS
# ============================================

# Employment Type
df['Employment-Type'] = (
    df['Employment-Type']
    .astype(str)
    .str.strip()
    .str.title()
)

# Education Level
df['Education_Level'] = (
    df['Education_Level']
    .astype(str)
    .str.strip()
    .str.title()
)

# City
df['city'] = (
    df['city']
    .astype(str)
    .str.strip()
    .str.lower()
)

df['city'] = df['city'].replace({
    'hyderbad': 'Hyderabad',
    'hyd': 'Hyderabad',
    'hyderabad': 'Hyderabad',
    'bangalore': 'Bengaluru',
    'bengaluru': 'Bengaluru'
})


# Customer Segment
df['customer_segment'] = (
    df['customer_segment']
    .astype(str)
    .str.strip()
    .str.title()
)


# Loan Type
df['loan_type'] = (
    df['loan_type']
    .astype(str)
    .str.strip()
    .str.lower()
)

df['loan_type'] = df['loan_type'].replace({
    'personal': 'Personal Loan',
    'personal loan': 'Personal Loan'
})


# Loan Approved
df['Loan_Approved'] = (
    df['Loan_Approved']
    .astype(str)
    .str.strip()
    .str.lower()
)

df['Loan_Approved'] = df['Loan_Approved'].replace({
    'yes': 'Yes',
    'y': 'Yes',
    '1': 'Yes',
    'no': 'No',
    'n': 'No',
    '0': 'No'
})


# ============================================
# 7. HANDLE INVALID VALUES
# ============================================

# Age must be between 0 and 100
df.loc[
    (df['Age'] < 0) | (df['Age'] > 100),
    'Age'
] = np.nan

# Credit Score must be between 300 and 850
df.loc[
    (df['credit_Score'] < 300) |
    (df['credit_Score'] > 850),
    'credit_Score'
] = np.nan

# Debt to Income must be between 0 and 1
df.loc[
    (df['debt_to_income'] < 0) |
    (df['debt_to_income'] > 1),
    'debt_to_income'
] = np.nan


# ============================================
# 8. HANDLE MISSING VALUES
# ============================================

numeric_columns = [
    'Age',
    'Annual_Income',
    'credit_Score',
    'Loan_Amount_INR',
    'Monthly_Visits',
    'debt_to_income'
]

# Fill numerical missing values with median
for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())


# Fill categorical missing values with mode
categorical_columns = [
    'Employment-Type',
    'Education_Level',
    'city',
    'customer_segment',
    'loan_type',
    'Loan_Approved'
]

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])


# ============================================
# 9. DETECT AND TREAT OUTLIERS
#    USING IQR METHOD
# ============================================

outlier_columns = [
    'Annual_Income',
    'Loan_Amount_INR'
]

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    # Cap outliers instead of deleting rows
    df[col] = df[col].clip(
        lower=lower,
        upper=upper
    )


# ============================================
# 10. FIX FINAL DATA TYPES
# ============================================

df['Age'] = df['Age'].astype(int)

df['Annual_Income'] = df['Annual_Income'].astype(float)

df['credit_Score'] = df['credit_Score'].astype(int)

df['Loan_Amount_INR'] = df['Loan_Amount_INR'].astype(float)

df['Monthly_Visits'] = df['Monthly_Visits'].astype(int)

df['debt_to_income'] = df['debt_to_income'].astype(float)


# ============================================
# 11. FINAL ERROR CHECK
# ============================================

print("\n========== FINAL DATASET INFO ==========")

print("\nShape:")
print(df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

print("\nUnique categorical values:")

for col in categorical_columns:
    print("\n", col)
    print(df[col].unique())


# ============================================
# 12. CREATE FINAL CLEAN DATASET
# ============================================

df_cleaned = df.copy()

print("\n========== CLEANING COMPLETED ==========")
print(df_cleaned.head())
```
