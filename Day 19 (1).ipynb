```python
#1.Class Name: Shape 
#Problem 
#Create an abstract class named Shape. 
#Explanation 
#A shape can have different formulas for area and perimeter. Therefore, create two abstract  methods: 
#• area()  
#• perimeter()  
#Create subclasses like Circle and Rectangle and implement both methods.


from abc import ABC, abstractmethod
import math
#Abstract class

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

# Circle class
class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius

#Rectangle class
class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

# Create objects
circle = Circle(5)
rectangle = Rectangle(10, 5)

# Display results
print("Circle Area:", circle.area())
print("Circle Perimeter:", circle.perimeter())

print("Rectangle Area:", rectangle.area())
print("Rectangle Perimeter:", rectangle.perimeter())
    
```

```python
#13.Class Name: Calculator
#Problem
#Create an abstract class named Calculator.
#Explanation
#Different calculators perform different operations.
#Create abstract method:
#• compute(a, b)
#Create subclasses:
#• AdditionCalculator
#• MultiplicationCalculator

class Calculator(ABC):

    @abstractmethod
    def compute(self, a, b):
        pass

class AdditionCalculator(Calculator):

    def compute(self, a, b):
        return a + b

class MultiplicationCalculator(Calculator):

    def compute(self, a, b):
        return a * b

# Create objects
addition = AdditionCalculator()
multiplication = MultiplicationCalculator()

# Calculate results
print("Addition:", addition.compute(10, 5))
print("Multiplication:", multiplication.compute(10, 5))
```

```python
#14.Class Name: Database
#Problem
#Create an abstract class named Database.
#Explanation
#Different databases connect and execute queries differently.
#Create abstract methods:
#• connect()
#• query(sql)
#Create subclasses:
#• MySQLDatabase
#• MongoDatabase


class Database(ABC):

    @abstractmethod
    def connect(self):
        pass

    @abstractmethod
    def query(self, sql):
        pass
class MySQLDatabase(Database):

    def connect(self):
        print("Connected to MySQL database.")

    def query(self, sql):
        print("Executing MySQL query:", sql)

class MongoDatabase(Database):

    def connect(self):
        print("Connected to MongoDB database.")

    def query(self, sql):
        print("Executing MongoDB query:", sql)

# Create objects
mysql = MySQLDatabase()
mongo = MongoDatabase()

# MySQL operations
mysql.connect()
mysql.query("SELECT * FROM students")

print()

# MongoDB operations
mongo.connect()
mongo.query("Find all students")
```

```python
#15.Class Name: Shape3D
#Problem
#Create an abstract class named Shape3D.
#Explanation
#Three-dimensional shapes have different formulas.
#Create abstract methods:
#• volume()
#• surface_area()
#Create subclasses:
#• Cube
#• Sphere

class Shape3D(ABC):

    @abstractmethod
    def volume(self):
        pass

    @abstractmethod
    def surface_area(self):
        pass

class Cube(Shape3D):

    def __init__(self, side):
        self.side = side

    def volume(self):
        return self.side ** 3

    def surface_area(self):
        return 6 * self.side ** 2

class Sphere(Shape3D):

    def __init__(self, radius):
        self.radius = radius

    def volume(self):
        return (4 / 3) * math.pi * self.radius ** 3

    def surface_area(self):
        return 4 * math.pi * self.radius ** 2

# Create objects
cube = Cube(5)
sphere = Sphere(3)

# Cube calculations
print("Cube Volume:", cube.volume())
print("Cube Surface Area:", cube.surface_area())

print()

# Sphere calculations
print("Sphere Volume:", sphere.volume())
print("Sphere Surface Area:", sphere.surface_area())
```

```python
#16. Class Name: Shape
#Problem
#Create an abstract class named Shape.
#Explanation
#The constructor should accept a shape's color.
#__init__(self, color)
#Create an abstract method:
#• area()
#Create subclass:
#• Circle
#Use super() to initialize the color.

class Shape(ABC):

    def __init__(self, color):
        self.color = color

    @abstractmethod
    def area(self):
        pass

class Circle(Shape):

    def __init__(self, color, radius):
        super().__init__(color)   # Initialize color from parent class
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

# Create Circle object
circle = Circle("Red", 5)

print("Color:", circle.color)
print("Area:", circle.area())
```

```python
#17.Class Name: Vehicle
#Problem
#Create an abstract class named Vehicle.
#Explanation
#Store the vehicle brand in the constructor.
#__init__(self, brand)
#Create abstract methods:
#• start()
#• stop()
#Create subclass:
#• Car
#Use super() to initialize the brand.

class Vehicle(ABC):

    def __init__(self, brand):
        self.brand = brand

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vehicle):

    def __init__(self, brand):
        super().__init__(brand)   # Initialize brand from parent class

    def start(self):
        print(self.brand, "car is starting.")

    def stop(self):
        print(self.brand, "car is stopping.")

# Create Car object
car = Car("Toyota")

print("Brand:", car.brand)
car.start()
car.stop()
```
