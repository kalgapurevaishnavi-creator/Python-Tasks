```python
#1.task.Create a hybrid inheritance structure with classes A → B and C → D (where D inherits
#from both B and C).
#Concept
#Hybrid Inheritance (Combination of Single + Multiple Inheritance)
#What to do
#• Create parent class A.
#• Create class B inheriting from A.
#• Create another class C.
#• Create class D inheriting from both B and C.
#How to solve
#1. Create class A.
#2. Add one method.
#3. Create class B(A).
#4. Create class C.
#5. Create class D(B, C).
#6. Create an object of D.
#7. Call methods from all parent classes.


#parent class
class A:
    def method_a(self):
        print("this is method A")

#class b inherits from A
class B(A):
    def method_b(self):
        print("this is method B")

#class c
class C:
    def method_c(self):
        print("this is method C")
        
#class D inherits from D
class D(B,C):
    def method_d(self):
        print("this is method D")

#create aobject
obj=D()
#call method from parent inherits class
obj.method_a()
obj.method_b()
obj.method_c()
obj.method_d()       
```

```python
#Task2.Create Person, Employee, Student, and TeachingAssistant using hybrid inheritance.
#Concept
#Real-world Hybrid Inheritance
#What to do
#• Person is the base class.
#• Employee inherits Person.
#• Student inherits Person.
#• TeachingAssistant inherits Employee and Student.
#How to solve
#Hierarchy
#Person
#/ \
#Employee Student
#\ /
#TeachingAssistant
#Steps
#1. Store name in Person.
#2. Employee stores salary.
#3. Student stores marks.
#4. TeachingAssistant stores department.
#5. Display all information.


#parent class
class Person:
    def __init__(self,name):
        self.name=name

#child class Employee
class Employee(Person):
    def __init__(self,name,salary):
        Person.__init__(self,name)
        self.salary=salary
        
#child class
class Student(Person):
    def __init__(self,name,marks):
        Person.__init__(self,name)
        self.marks=marks

#teachingassistance class
class TeachingAssistant(Employee,Person):
    def __init__(self,name,salary,marks,deparment):
        Employee.__init__(self,name,salary)
        self.marks=marks
        self.deparment=deparment

   #display all information
    def display(self):
       print("Name:",self.name)
       print("Salary:",self.salary)
       print("Marks:",self.marks)
       print("Deparment:",self.deparment)

#create a object
ta=TeachingAssistant("vaishu",10000,500,"DaTa science")

#dispaly all information
ta.display()       
```

```python
#3.Create Vehicle → Car, Electric → ElectricCar (ElectricCar inherits from both).
#Concept
#Hybrid Inheritance
#What to do
#• Vehicle is parent.
#• Car inherits Vehicle.
#• Electric is another parent.
#• ElectricCar inherits Car and Electric.
#How to solve
#Hierarchy
#Vehicle
#|
#Car Electric
#\ /
#ElectricCar

#Steps
#1. Vehicle has start().
#2. Car has drive().
#3. Electric has battery().
#4. ElectricCar accesses all methods.


#parent class
class Vehicle:
    def start(self):
        print("vehicle is starting")

#class car inherits vehicle
class Car(Vehicle):
    def Drive(self):
        print("car is driving")

#another parent class
class Electric:
    def battery(self):
        print("battery is fully charged")
        
#class ElectricCar inherits car and Electric
class ElectricCar(Car,Electric):
    def display(self):
        print("Electric car")


#craete a object        
ec=ElectricCar()

#display all information
ec.start()
ec.Drive()
ec.battery()
ec.display()
```

```python
#4.Demonstrate Hybrid Inheritance and print Method Resolution Order (MRO).
#Concept
#Method Resolution Order (MRO)
#What to do
#Create
#A
#|
#B
#C
#D(B,C)
#Print
#D.mro()
#How to solve
#1. Create four classes.
#2. Give same method name in multiple classes.
#3. Print ClassName.mro().
#4. Observe Python's searching order.



#parent class
class A:
    def display(self):
        print("display A")

#child class
class B(A):
    def display(self):
        print("display B")

#another parent class
class C:
    def display(self):
        print("display C")

#class D inherits from B and C
class D(B,C):
    pass

#create object
d=D()
#display all information
d.display() 
#print method resolution order (MRO)
print(D.mro())
```

```python
#5.Create class A, B, C, D and solve the Diamond Problem using super().
#Concept
#Diamond Problem
#Hierarchy
#A
#/ \
#B C
#\ /
#D
#What to do
#Use super() everywhere.
#How to solve
#1. Create constructor in A
#2. Call super() in B.
#3. Call super() in C.
#4. Call super() in D.
#5. Create object of D.
#6. Observe A constructor runs only once.

#child class A
class A:
    def __init__(self):
        print("constructor A")
        
#child class B
class B(A):
    #constructor
    def __init__(self):
        super().__init__()
        print("constructor B")
        
# child class C
class C(A):
    #constructor
    def __init__(self):
        super().__init__()
        print("constructor C")

# child class D
class D(B,C):
   #constructor
    def __init__(self):
        super().__init__()
        print("constructor D")

#create a object
d=D()
#display all informaton
print(D.mro())
```

```python
#6.Create a banking system using hybrid inheritance. Concept 
#Hybrid + Real-world Example 
#What to do 
#Account 
# | 
#Savings 
#Loan 
#SavingsLoanAccount
#How to solve 
#1. Account stores account number.  
#2. Savings stores balance.  
#3. Loan stores loan amount.  
#4. SavingsLoanAccount inherits both. 
#5. Print all details.  


#Parent class
class Account:
    def __init__(self, acc_no):
        self.acc_no = acc_no


# Child class 1
class Saving(Account):
    def __init__(self, acc_no, balance):
        Account.__init__(self, acc_no)
        self.balance = balance


# Child class 2
class Loan:
    def __init__(self, loan_amount):
        self.loan_amount = loan_amount


# Hybrid Inheritance
class SavingLoanAccount(Saving, Loan):
    def __init__(self, acc_no, balance, loan_amount):
        Saving.__init__(self, acc_no, balance)
        Loan.__init__(self, loan_amount)

    def display(self):
        print("Account Number:", self.acc_no)
        print("Balance:", self.balance)
        print("Loan Amount:", self.loan_amount)


# Create object
obj = SavingLoanAccount(101, 50000, 20000)
obj.display()
```

```python
#7.Create Shape, Color, Circle, and ColoredCircle. Concept 
#Hybrid Inheritance 
#What to do 
#Shape 
# | 
#Circle 
#Color 
#ColoredCircle 
#How to solve 
#1. Shape has area().  
#2. Color has color().  
#3. Circle inherits shape.
#4. ColoredCircle inherits Circle and Color. 
#5. Display area and color.  


# Parent class
class Shape:
    def area(self, radius):
        return 3.14 * radius * radius

# Parent class
class Color:
    def color(self, color_name):
        print("Color:", color_name)

# Child class
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

# Hybrid Inheritance
class ColoredCircle(Circle, Color):
    def __init__(self, radius, color_name):
        Circle.__init__(self, radius)
        self.color_name = color_name

    def display(self):
        print("Area:", self.area(self.radius))
        self.color(self.color_name)


# Create object
obj = ColoredCircle(3, "perple")
obj.display()
```

```python
#8.Create constructors at every level. Concept
#Constructor Chaining #What to do #Every class has its own constructor. #How to solve 
#1. Create constructor in every class.
#2. Use super() everywhere. 
#3. Print constructor execution order.
#4. Observe execution using MRO.


# Parent class
class A:
    def __init__(self):
        super().__init__()
        print("Constructor A")


# Child class B
class B(A):
    def __init__(self):
        super().__init__()
        print("Constructor B")


# Child class C
class C(B):
    def __init__(self):
        super().__init__()
        print("Constructor C")

#child class D
class D(C):
    def __init__(self):
        super().__init__()
        print("Constructor D")


# Create object
obj = D()

# Print MRO
print("\nMethod Resolution Order (MRO):")
print(D.mro())
```

```python
#15.Create Circle and Rectangle both having area().
#Concept
#Duck Typing
#What to do
#Function
#calculate(obj)
#How to solve
#1. Circle implements area().
#2. Rectangle implements area().
#3. Function calls area().
#4. Print returned value.


import math
# Circle class
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius * self.radius

# Rectangle class
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

# Function that accepts any object
def calculate(obj):
    print("Area:", obj.area())


# Create objects
circle = Circle(5)
rectangle = Rectangle(10, 4)

# Call function
calculate(circle)
calculate(rectangle)
```

```python
#19.Create a hybrid inheritance structure and apply Duck Typing. Concept 
#Hybrid + Duck Typing 
#What to do 
#Create hybrid inheritance. 
#Create another unrelated class with same method. 
#How to solve 
#1. Create hybrid hierarchy.  
#2. Final child has method show().  
#3. Another unrelated class also has show().  
#4. One function calls show() for both. 

#Parent class
class A:
    pass

# Child class B
class B(A):
    pass

# Child class C
class C(A):
    pass

# Final child class (Hybrid Inheritance)
class D(B, C):
    def show(self):
        print("Show method from Hybrid Child D")

# Unrelated class
class Display:
    def show(self):
        print("Show method from Display class")

# Function using Duck Typing
def display_object(obj):
    obj.show()

# Create objects
d = D()
display = Display()

# Call function
display_object(d)
display_object(display)
```

```python
#20.Create Employee hierarchy using Hybrid Inheritance and use Duck Typing. Concept 
#Hybrid Inheritance + Duck Typing 
#What to do 
#Hierarchy 
#Person 
# | 
#Employee 
#Manager 
#HR 
#TeamLead 
#Another class 
#Freelancer 
#Both TeamLead and Freelancer have 
#work() 
#How to solve
#1 Build hybrid inheritance.  
#2. Final child defines work().  
#3. Freelancer also defines work().  
#4. Write a function:  
#def do_work(obj): 
# obj.work() 
#5. Pass both objects to the function.  
#6. Observe that the same function works for unrelated classes because of Duck Typing.


# Parent class
class Person:
    def show_person(self):
        print("Person class")

# Child class
class Employee(Person):
    def show_employee(self):
        print("Employee class")

# Child class 1
class Manager(Employee):
    def show_manager(self):
        print("Manager class")

# Child class 2
class HR(Employee):
    def show_hr(self):
        print("HR class")

# Final child class (Hybrid Inheritance)
class TeamLead(Manager, HR):
    def work(self):
        print("TeamLeead is managing the team.")

# Unrelated class
class Freelancer:
    def work(self):
        print("Freelancer is completing the project.")

# Duck Typing function
def do_work(obj):
    obj.work()

# Create objects
teamlead = TeamLead()
freelancer = Freelancer()

# Pass objects to function
do_work(teamlead)
do_work(freelancer)
```
